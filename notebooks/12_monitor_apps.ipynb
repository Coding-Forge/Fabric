```python
"""Embedded Monitor implementation; generated from tested repository sources."""

import hashlib
import json
import math
import re
from datetime import datetime, timezone
from pathlib import Path, PurePosixPath
from uuid import UUID

ITEM_TYPES = {"reports": "Report", "datasets": "SemanticModel", "dataflows": "Dataflow"}


def _text(value, field, required=False):
    if value is None and not required:
        return None
    if not isinstance(value, str) or not value.strip():
        raise ValueError(f"Invalid lineage field: {field}")
    return value.strip()


def _identifier(value, field, required=False):
    text = _text(value, field, required)
    return text.lower() if text else None


def _records(parent, field, required=False):
    value = parent.get(field)
    if value is None and not required:
        return []
    if not isinstance(value, list) or any(not isinstance(row, dict) for row in value):
        raise ValueError(f"Invalid lineage collection: {field}")
    return value


def scan_snapshot(
    scan, scan_id, requested_workspace_ids, observed_at_utc, full_scan=False
):
    """Build a batch observation, not an exhaustive or historical tenant graph.

    Unknown fields are ignored. Malformed recognized fields raise ``ValueError``
    rather than becoming successful empty observations. Missing optional
    metadata and references outside this batch remain unresolved.
    """
    if not isinstance(scan_id, str) or not re.fullmatch(
        r"[A-Za-z0-9][A-Za-z0-9_-]{0,127}", scan_id
    ):
        raise ValueError("Invalid lineage scan ID")
    if not isinstance(scan, dict) or any(key in scan for key in ("ERROR", "error")):
        raise ValueError("Invalid lineage scan response")
    requested = sorted(
        {
            _identifier(value, "requestedWorkspaceId", required=True)
            for value in requested_workspace_ids
        }
    )
    workspaces = _records(scan, "workspaces", required=True)
    returned = set()
    items = {}
    raw_items = {}
    for workspace in workspaces:
        ws_id = _identifier(workspace.get("id"), "workspace.id", required=True)
        if ws_id in returned or ws_id not in requested:
            raise ValueError("Duplicate or unexpected lineage workspace")
        if any(key in workspace for key in ("ERROR", "error", "scanError")):
            raise ValueError("Failed lineage workspace")
        returned.add(ws_id)
        for key, kind in ITEM_TYPES.items():
            for raw in _records(workspace, key):
                item_id = _identifier(
                    raw.get("objectId") if key == "dataflows" else raw.get("id"),
                    f"{key}.id",
                    required=True,
                )
                if item_id in items:
                    raise ValueError("Duplicate lineage item")
                items[item_id] = {
                    "id": item_id,
                    "name": _text(raw.get("name"), f"{key}.name"),
                    "type": kind,
                    "workspaceId": ws_id,
                }
                raw_items[item_id] = raw

    datasources = {}
    for raw in _records(scan, "datasourceInstances"):
        source_id = _identifier(raw.get("datasourceId"), "datasourceId", required=True)
        if source_id in datasources:
            raise ValueError("Duplicate lineage datasource")
        kind = _text(raw.get("datasourceType"), "datasourceType")
        details = raw.get("connectionDetails")
        if details is not None and not isinstance(details, dict):
            raise ValueError("Invalid lineage connectionDetails")
        # Do not export arbitrary connection strings, URLs or expression payloads.
        labels = [
            _text((details or {}).get(key), f"connectionDetails.{key}")
            for key in ("server", "database")
        ]
        label = " / ".join(part for part in labels if part)
        datasources[source_id] = {
            "id": source_id,
            "name": f"{kind}: {label}" if kind and label else (kind or label or None),
            "type": "DataSource",
            "workspaceId": None,
        }

    edges = {}

    def add(query_id, related_id, relation, related_type=None, related_ws=None):
        query = items[query_id]
        related = (
            datasources.get(related_id)
            if related_type == "DataSource"
            else items.get(related_id)
        )
        if related is None:
            related = {
                "id": related_id,
                "name": None,
                "type": related_type,
                "workspaceId": related_ws,
            }

        def emit(source, target, direction):
            row = {
                "QueryItemId": source["id"],
                "QueryItemName": source["name"],
                "QueryItemType": source["type"],
                "QueryWorkspaceId": source["workspaceId"],
                "Direction": direction,
                "RelatedItemId": target["id"],
                "RelatedItemName": target["name"],
                "RelatedItemType": target["type"],
                "RelatedWorkspaceId": target["workspaceId"],
                "RelationType": relation,
                "Source": "Scanner",
            }
            key = (source["id"], direction, target["id"], relation)
            existing = edges.get(key)
            if existing is not None:
                for field, value in row.items():
                    if existing[field] is None:
                        existing[field] = value
                    elif value is not None and existing[field] != value:
                        raise ValueError("Conflicting lineage reference metadata")
            else:
                edges[key] = row

        emit(query, related, "upstream")
        if related_type != "DataSource" and related_id in items:
            emit(related, query, "downstream")

    for item_id, raw in raw_items.items():
        if items[item_id]["type"] == "Report" and raw.get("datasetId") is not None:
            add(
                item_id,
                _identifier(raw["datasetId"], "datasetId", True),
                "ReportModel",
                "SemanticModel",
            )
        for field, target, relation, kind in (
            ("upstreamDatasets", "targetDatasetId", "UpstreamDataset", "SemanticModel"),
            ("upstreamDataflows", "targetDataflowId", "UpstreamDataflow", "Dataflow"),
        ):
            for ref in _records(raw, field):
                add(
                    item_id,
                    _identifier(ref.get(target), target, True),
                    relation,
                    kind,
                    _identifier(ref.get("groupId"), "groupId"),
                )
        for ref in _records(raw, "relations"):
            add(
                item_id,
                _identifier(
                    ref.get("dependentOnArtifactId"), "dependentOnArtifactId", True
                ),
                _text(ref.get("relationType"), "relationType") or "Relation",
            )
        for ref in _records(raw, "datasourceUsages"):
            add(
                item_id,
                _identifier(
                    ref.get("datasourceInstanceId"), "datasourceInstanceId", True
                ),
                "DataSource",
                "DataSource",
            )

    return {
        "schemaVersion": 1,
        "source": "Scanner",
        "scanId": scan_id,
        "observedAtUtc": observed_at_utc,
        "scanMode": "full" if full_scan else "incremental",
        "requestedWorkspaceIds": requested,
        "returnedWorkspaceIds": sorted(returned),
        "items": [items[key] for key in sorted(items)],
        "complete": True,
        "edges": [edges[key] for key in sorted(edges)],
    }


SCHEMA_VERSION = 1
LAKEHOUSE_FILES = "/lakehouse/default/Files"
PERMISSIONS = {
    "fabricitemaccesspermissions": "Scoped",
    "fabricitemaccesspermissions_tenant": "Tenant",
}
MODULES = (
    *PERMISSIONS,
    "fabricitemjobhistory",
    "fabricitemdefinitions",
    "fabricconnections",
    "fabricdeploymentpipelineoperations",
)
CORE_SOURCE_PREFIXES = {*MODULES, "items", "workspaces", "catalog", "lineage_edges"}
POWER_BI = {"Report", "Dashboard", "SemanticModel", "App", "Dataflow"}
COMMON = {
    "schema_version": "int",
    "observation_id": "string",
    "source_path": "string",
    "source_sha256": "string",
    "record_sha256": "string",
    "observed_at_utc": "string",
    "module": "string",
    "mode": "string",
    "run_id": "string",
    "cycle_id": "string",
    "coverage_status": "string",
}
TARGET_COLUMNS = {
    "workspace_id": "string",
    "item_id": "string",
    "item_type": "string",
}
SCHEMAS = {
    "collection_health": {
        "complete": "boolean",
        "finished": "boolean",
        "successful_resources": "long",
        "failed_resources": "long",
        "record_count": "long",
        "coverage": "string",
        "issue": "string",
    },
    "inventory": {
        **TARGET_COLUMNS,
        "item_name": "string",
        "inventory_source": "string",
    },
    "permission_cycles": {
        "chain_valid": "boolean",
        "cycle_finished": "boolean",
        "cycle_healthy": "boolean",
        "slice_count": "long",
        "selected_items": "long",
    },
    "permission_resources": {
        **TARGET_COLUMNS,
        "observation_unit": "string",
        "record_count": "long",
        "resource_outcome": "string",
    },
    "permission_grants": {
        **TARGET_COLUMNS,
        "principal_id": "string",
        "principal_type": "string",
        "permissions_json": "string",
        "additional_permissions_json": "string",
    },
    "job_resources": {**TARGET_COLUMNS, "record_count": "long"},
    "job_observations": {
        **TARGET_COLUMNS,
        "job_id": "string",
        "job_type": "string",
        "invoke_type": "string",
        "status": "string",
        "start_time_utc": "string",
        "end_time_utc": "string",
        "duration_seconds": "double",
        "root_activity_id": "string",
        "failure_code": "string",
    },
    "lineage_resources": {
        **TARGET_COLUMNS,
        "evidence_type": "string",
        "record_count": "long",
        "diagnostic_count": "long",
    },
    "lineage_edges": {
        **TARGET_COLUMNS,
        "related_workspace_id": "string",
        "related_item_id": "string",
        "related_type": "string",
        "direction": "string",
        "relation_type": "string",
        "reference": "string",
        "resolution": "string",
        "evidence_type": "string",
        "runtime_verified": "boolean",
    },
    "lineage_diagnostics": {
        **TARGET_COLUMNS,
        "reason": "string",
        "location": "string",
    },
}
CORE_DATASETS = tuple(SCHEMAS)
ESTATE_SCHEMAS = {
    "estate_coverage": {
        "analysis": "string",
        "source_date": "string",
        "record_count": "long",
        "issue": "string",
    },
    "activity_events": {
        **TARGET_COLUMNS,
        "event_id": "string",
        "event_time_utc": "string",
        "operation": "string",
        "workload": "string",
        "actor_key": "string",
        "actor_type": "string",
        "is_success": "boolean",
        "source_date": "string",
    },
    "catalog_workspaces": {
        "workspace_id": "string",
        "workspace_name": "string",
        "workspace_type": "string",
        "state": "string",
        "capacity_id": "string",
        "returned_collections_json": "string",
        "source_date": "string",
    },
    "catalog_items": {
        **TARGET_COLUMNS,
        "item_name": "string",
        "source_date": "string",
    },
    "catalog_models": {
        **TARGET_COLUMNS,
        "model_name": "string",
        "schema_coverage": "string",
        "table_count": "long",
        "relationship_count": "long",
        "source_date": "string",
    },
    "model_tables": {
        **TARGET_COLUMNS,
        "table_name": "string",
        "is_hidden": "boolean",
        "column_count": "long",
        "measure_count": "long",
        "source_date": "string",
    },
    "model_columns": {
        **TARGET_COLUMNS,
        "table_name": "string",
        "column_name": "string",
        "data_type": "string",
        "is_hidden": "boolean",
        "source_date": "string",
    },
    "model_measures": {
        **TARGET_COLUMNS,
        "table_name": "string",
        "measure_name": "string",
        "data_type": "string",
        "expression_returned": "boolean",
        "source_date": "string",
    },
    "model_relationships": {
        **TARGET_COLUMNS,
        "from_table": "string",
        "from_column": "string",
        "to_table": "string",
        "to_column": "string",
        "from_cardinality": "string",
        "to_cardinality": "string",
        "is_active": "boolean",
        "source_date": "string",
    },
    "connection_observations": {
        "connection_id": "string",
        "connection_name": "string",
        "gateway_id": "string",
        "connectivity_type": "string",
        "privacy_level": "string",
        "connection_type": "string",
    },
    "gateway_observations": {
        "gateway_id": "string",
        "gateway_name": "string",
        "gateway_type": "string",
        "source_date": "string",
    },
    "gateway_datasources": {
        "gateway_id": "string",
        "datasource_id": "string",
        "datasource_name": "string",
        "datasource_type": "string",
        "detail_returned": "boolean",
        "source_date": "string",
    },
    "gateway_status": {
        "gateway_id": "string",
        "datasource_id": "string",
        "status_code": "string",
        "http_status": "long",
        "connectivity_observation": "string",
        "source_date": "string",
    },
    "refresh_observations": {
        **TARGET_COLUMNS,
        "dataset_name": "string",
        "refresh_id": "string",
        "refresh_type": "string",
        "status": "string",
        "start_time_utc": "string",
        "end_time_utc": "string",
        "duration_seconds": "double",
        "failure_code": "string",
        "source_date": "string",
    },
    "refreshable_observations": {
        **TARGET_COLUMNS,
        "dataset_name": "string",
        "capacity_id": "string",
        "refresh_count": "long",
        "refresh_failures": "long",
        "schedule_enabled": "boolean",
        "source_date": "string",
    },
}
SCHEMAS.update(ESTATE_SCHEMAS)
ESTATE_DATASETS = tuple(SCHEMAS)
ESTATE_PREFIXES = {"activity", "gateways", "datasetrefresh", "datasetrefreshable"}
MODULE_PREFIXES = {
    "capacity",
    "capacity_metrics",
    "graph",
    "roles",
    "tenant",
    "delegatedsettings",
    "domains",
    "gitconnections",
    "tags",
    "widelysharedlinks",
}
MODULE_SCHEMAS = {
    "apps": {
        "app_id": "string",
        "app_name": "string",
        "workspace_id": "string",
        "last_update_utc": "string",
        "source_date": "string",
    },
    "capacities": {
        "capacity_id": "string",
        "capacity_name": "string",
        "sku": "string",
        "state": "string",
        "region": "string",
        "source_date": "string",
    },
    "capacity_metric_observations": {
        "capacity_id": "string",
        "sku": "string",
        "metric_group": "string",
        "metric_date": "string",
        "timepoint_raw": "string",
        "timepoint_utc": "string",
        "metric_name": "string",
        "metric_value": "double",
        "source_date": "string",
    },
    "directory_users": {
        "principal_id": "string",
        "account_enabled": "boolean",
        "user_type": "string",
        "assigned_license_count": "long",
        "source_date": "string",
    },
    "directory_groups": {
        "group_id": "string",
        "security_enabled": "boolean",
        "mail_enabled": "boolean",
        "group_types_json": "string",
        "is_audience_group": "boolean",
        "source_date": "string",
    },
    "group_memberships": {
        "group_id": "string",
        "principal_id": "string",
        "membership_kind": "string",
        "is_audience_group": "boolean",
        "source_date": "string",
    },
    "user_licenses": {
        "principal_id": "string",
        "sku_id": "string",
        "disabled_plans_json": "string",
        "source_date": "string",
    },
    "subscribed_skus": {
        "sku_id": "string",
        "sku_part_number": "string",
        "capability_status": "string",
        "consumed_units": "long",
        "enabled_units": "long",
        "suspended_units": "long",
        "warning_units": "long",
        "source_date": "string",
    },
    "workspace_role_scopes": {
        "workspace_id": "string",
        "roles_returned": "boolean",
        "assignment_count": "long",
        "source_date": "string",
    },
    "workspace_role_assignments": {
        "workspace_id": "string",
        "principal_id": "string",
        "principal_type": "string",
        "workspace_role": "string",
        "source_date": "string",
    },
    "tenant_settings": {
        "setting_name": "string",
        "title": "string",
        "setting_group": "string",
        "enabled": "boolean",
        "can_specify_groups": "boolean",
        "enabled_group_count": "long",
        "excluded_group_count": "long",
        "source_date": "string",
    },
    "delegated_setting_scopes": {
        "scope_type": "string",
        "entity_id": "string",
        "settings_returned": "boolean",
        "setting_count": "long",
        "source_date": "string",
    },
    "delegated_setting_overrides": {
        "scope_type": "string",
        "entity_id": "string",
        "setting_name": "string",
        "title": "string",
        "setting_group": "string",
        "enabled": "boolean",
        "delegated_from": "string",
        "enabled_group_count": "long",
        "excluded_group_count": "long",
        "source_date": "string",
    },
    "tenant_setting_groups": {
        "scope_type": "string",
        "entity_id": "string",
        "setting_name": "string",
        "group_id": "string",
        "group_rule": "string",
        "source_date": "string",
    },
    "domains": {
        "domain_id": "string",
        "domain_name": "string",
        "parent_domain_id": "string",
        "source_date": "string",
    },
    "domain_workspaces": {
        "domain_id": "string",
        "workspace_id": "string",
        "workspace_name": "string",
        "source_date": "string",
    },
    "workspace_git_connections": {
        "workspace_id": "string",
        "provider_type": "string",
        "owner_key": "string",
        "organization_name": "string",
        "project_name": "string",
        "repository_name": "string",
        "branch_name": "string",
        "directory_key": "string",
        "connection_state": "string",
        "last_sync_utc": "string",
        "source_date": "string",
    },
    "tags": {
        "tag_id": "string",
        "tag_name": "string",
        "scope_type": "string",
        "domain_id": "string",
        "source_date": "string",
    },
    "organization_shares": {
        "artifact_id": "string",
        "artifact_type": "string",
        "access_right": "string",
        "share_type": "string",
        "source_date": "string",
    },
    "publish_to_web_shares": {
        "artifact_id": "string",
        "artifact_type": "string",
        "access_right": "string",
        "share_type": "string",
        "source_date": "string",
    },
    "deployment_pipelines": {
        "pipeline_id": "string",
        "pipeline_name": "string",
        "source_date": "string",
    },
    "deployment_operation_resources": {
        "pipeline_id": "string",
        "record_count": "long",
        "resource_outcome": "string",
        "source_date": "string",
    },
    "deployment_operations": {
        "pipeline_id": "string",
        "operation_id": "string",
        "operation_type": "string",
        "status": "string",
        "last_updated_utc": "string",
        "execution_start_utc": "string",
        "execution_end_utc": "string",
        "duration_seconds": "double",
        "source_stage_id": "string",
        "target_stage_id": "string",
        "source_date": "string",
    },
}
SCHEMAS.update(MODULE_SCHEMAS)


def canonical(value):
    return json.dumps(value, sort_keys=True, separators=(",", ":"), allow_nan=False)


def sha256(data):
    return hashlib.sha256(data).hexdigest()


def utc(value):
    if not isinstance(value, str):
        raise TypeError("Invalid UTC timestamp")
    parsed = datetime.fromisoformat(value.replace("Z", "+00:00"))
    if parsed.tzinfo is None:
        raise ValueError("Naive timestamp")
    return (
        parsed.astimezone(timezone.utc)
        .isoformat(timespec="microseconds")
        .replace("+00:00", "Z")
    )


def text(value):
    if not isinstance(value, str) or not value.strip():
        raise ValueError("Missing text")
    return value


def uuid(value):
    if not isinstance(value, str) or str(UUID(value)) != value:
        raise ValueError("Invalid canonical identity")
    return value


def records(value):
    if not isinstance(value, list) or any(not isinstance(v, dict) for v in value):
        raise ValueError("Invalid record collection")
    return value


def target(parent):
    return uuid(parent["workspaceId"]), uuid(parent["itemId"]), text(parent["type"])


def target_fields(parent):
    return dict(zip(TARGET_COLUMNS, target(parent), strict=True))


def document(path, data):
    """Parse bytes without propagating source text into diagnostics."""
    relative = PurePosixPath(path)
    if relative.is_absolute() or ".." in relative.parts or "\\" in path:
        raise ValueError("Unsafe source path")
    try:
        value = json.loads(data)
        canonical(value)
        issue = None
    except (ValueError, UnicodeError, RecursionError):
        value, issue = None, "malformed_json"
    return {
        "path": path,
        "sha256": sha256(data),
        "value": value,
        "issue": issue,
    }


def load_documents(
    raw_root,
    *,
    max_files=50000,
    max_bytes=128 * 1024 * 1024,
    max_file_bytes=8 * 1024 * 1024,
    source_prefixes=None,
    allow_empty=False,
):
    """Bounded driver-side reads of the attached Lakehouse; never write raw files."""
    root = Path(raw_root)
    allowed = Path(LAKEHOUSE_FILES)
    if (
        not root.is_absolute()
        or ".." in root.parts
        or root.resolve() == allowed
        or allowed not in root.resolve().parents
        or not root.is_dir()
    ):
        raise ValueError(
            "Select an existing folder inside the attached Lakehouse Files"
        )
    result, total = [], 0
    prefixes = CORE_SOURCE_PREFIXES | ESTATE_PREFIXES | MODULE_PREFIXES
    if source_prefixes is not None:
        selected = set(source_prefixes)
        if not selected or not selected.issubset(prefixes):
            raise ValueError("Select known source folders")
        prefixes = selected
    for path in sorted(root.rglob("*.json")):
        relative = path.relative_to(root)
        if relative.parts[0] not in prefixes or path.name in {
            "request_gate.json",
            "progress.json",
            "tenant_progress.json",
        }:
            continue
        if relative.parts[0] == "gateways" and path.name == "users.json":
            continue
        if path.is_symlink() or root.resolve() not in path.resolve().parents:
            raise ValueError("Source path escaped the raw folder")
        size = path.stat().st_size
        total += size
        if len(result) >= max_files or total > max_bytes or size > max_file_bytes:
            raise ValueError("Raw evidence exceeds configured driver safety limits")
        data = path.read_bytes()
        if len(data) != size:
            raise ValueError("Source changed during reading")
        result.append(document(relative.as_posix(), data))
    if not result and not allow_empty:
        raise ValueError("No supported collection files; refusing an empty import")
    return result


def row(table, doc, key, *, observed=None, **values):
    path = doc["path"]
    parts = PurePosixPath(path).parts
    value = doc["value"] if isinstance(doc["value"], dict) else {}
    timestamp = observed if observed is not None else value.get("observedAt")
    try:
        timestamp = utc(timestamp)
    except (ValueError, TypeError):
        timestamp = None
    result = dict.fromkeys({**COMMON, **SCHEMAS[table]})
    result.update(
        schema_version=SCHEMA_VERSION,
        observation_id=sha256(canonical([table, path, key]).encode()),
        source_path=path,
        source_sha256=doc["sha256"],
        observed_at_utc=timestamp,
        module=parts[0],
        mode=PERMISSIONS.get(parts[0]),
        run_id=next(
            (
                value[key]
                for key in ("runId", "scanId")
                if isinstance(value.get(key), str)
            ),
            None,
        ),
        coverage_status="success",
    )
    if set(values) - set(result):
        raise ValueError("Unknown contract column")
    result.update(values)
    if parts[0] in {"items", "workspaces", "catalog"} or result["coverage_status"] in {
        "invalid",
        "unfinished",
        "legacy_unverified",
    }:
        result["observation_id"] = sha256(
            canonical([table, path, key, doc["sha256"]]).encode()
        )
    return result


def _validate_records(module, resource):
    parent = resource["parent"]
    inventory = parent.get("observationUnit") == "inventoryPage"
    if module in PERMISSIONS:
        if inventory:
            if not isinstance(parent.get("inventoryComplete"), bool):
                raise ValueError("Invalid inventory completion")
            if PERMISSIONS[module] == "Scoped":
                uuid(parent["workspaceId"])
        else:
            target(parent)
    elif module == "fabricitemjobhistory":
        if target(parent)[2] not in {"Notebook", "DataPipeline", "SparkJobDefinition"}:
            raise ValueError("Unsupported job item")
    elif (
        module == "fabricitemdefinitions"
        and parent
        and target(parent)[2] not in {"Notebook", "DataPipeline"}
    ):
        raise ValueError("Unsupported definition item")
    if not resource["complete"]:
        text(resource["error"])
        if "records" in resource or (inventory and parent["inventoryComplete"]):
            raise ValueError("Failed resource contains records or complete inventory")
        return
    if "error" in resource:
        raise ValueError("Successful resource contains an error")
    values = records(resource["records"])
    identities = []
    for value in values:
        if module in PERMISSIONS and inventory:
            identity = target(value)
            if PERMISSIONS[module] == "Scoped" and identity[0] != parent["workspaceId"]:
                raise ValueError("Inventory workspace mismatch")
            identities.append(
                identity[:2] if PERMISSIONS[module] == "Tenant" else identity
            )
        elif module in PERMISSIONS:
            principal, detail = value["principal"], value["itemAccessDetails"]
            text(principal["id"])
            if not isinstance(principal["type"], str):
                raise ValueError("Invalid principal type")
            if not (
                detail["type"] == parent["type"]
                or (parent["type"] in POWER_BI and detail["type"] is None)
            ):
                raise ValueError("Permission item type mismatch")
            for field in ("permissions", "additionalPermissions"):
                if field == "additionalPermissions" and (
                    field not in detail
                    or (detail[field] is None and parent["type"] in POWER_BI)
                ):
                    continue
                if not isinstance(detail[field], list) or any(
                    not isinstance(v, str) for v in detail[field]
                ):
                    raise ValueError("Invalid permission list")
        elif module == "fabricitemjobhistory":
            for field in ("id", "jobType", "invokeType", "status"):
                text(value[field])
            if value["itemId"] != parent["itemId"]:
                raise ValueError("Job parent mismatch")
            for field in ("startTimeUtc", "endTimeUtc"):
                if value.get(field) is not None:
                    utc(value[field])
            if value.get("failureReason") is not None and not isinstance(
                value["failureReason"], dict
            ):
                raise ValueError("Invalid job failure metadata")
            if value.get("rootActivityId") is not None and not isinstance(
                value["rootActivityId"], str
            ):
                raise ValueError("Invalid activity identity")
            identities.append(value["id"])
    if len(identities) != len(set(identities)):
        raise ValueError("Duplicate observation identity")


def _resource_name_endpoint(module, resource):
    parent = resource["parent"]
    if module in PERMISSIONS:
        if parent.get("observationUnit") == "inventoryPage":
            if PERMISSIONS[module] == "Tenant":
                return "inventory.json", "admin/items?state=active"
            workspace = parent["workspaceId"]
            return f"inventory_{workspace}.json", f"admin/items?workspaceId={workspace}"
        workspace, item, kind = target(parent)
        suffix = f"?type={kind}" if kind in POWER_BI else ""
        return (
            f"{workspace}_{item}.json",
            f"admin/workspaces/{workspace}/items/{item}/users{suffix}",
        )
    if module == "fabricitemjobhistory":
        workspace, item, _ = target(parent)
        return (
            f"{workspace}_{item}.json",
            f"workspaces/{workspace}/items/{item}/jobs/instances",
        )
    if module == "fabricitemdefinitions":
        if not parent:
            return "inventory.json", "admin/items"
        workspace, item, kind = target(parent)
        endpoint = (
            f"notebooks/{item}/getDefinition?format=ipynb"
            if kind == "Notebook"
            else f"dataPipelines/{item}/getDefinition"
        )
        return f"{workspace}_{item}.json", f"workspaces/{workspace}/{endpoint}"
    return None, None


def build_runs(documents):
    grouped = {}
    for doc in documents:
        path = PurePosixPath(doc["path"])
        if path.parts[0] in MODULES:
            grouped.setdefault(str(path.parent), {})[path.name] = doc
    result = {}
    for folder, files in sorted(grouped.items()):
        module = folder.split("/")[0]
        manifest_doc = files.get("manifest.json")
        representative = manifest_doc or files[min(files)]
        run = {
            "doc": representative,
            "module": module,
            "manifest": None,
            "resources": [],
            "valid": False,
            "issue": None,
            "chain": None,
        }
        result[folder + "/manifest.json"] = run
        try:
            parts = folder.split("/")
            if len(parts) != 5 or not re.fullmatch(r"[a-f0-9]{32}", parts[-1]):
                raise ValueError("Invalid immutable run path")
            datetime.strptime("/".join(parts[1:4]), "%Y/%m/%d")
            if manifest_doc is None or manifest_doc["issue"]:
                raise ValueError("Missing or malformed manifest")
            manifest = manifest_doc["value"]
            if not isinstance(manifest, dict):
                raise TypeError("Invalid manifest object")
            run["manifest"] = manifest
            if (
                manifest["runId"] != parts[-1]
                or not isinstance(manifest["complete"], bool)
                or not isinstance(manifest["finished"], bool)
                or not isinstance(manifest["errors"], list)
            ):
                raise ValueError("Invalid manifest fields")
            final_time = utc(manifest["observedAt"])
            entries = records(manifest["resources"])
            names = []
            if not manifest["finished"]:
                if manifest["complete"] or entries or manifest["errors"]:
                    raise ValueError("Invalid initial manifest")
                run["valid"] = True
                continue
            for entry in entries:
                name = text(entry["file"])
                if (
                    PurePosixPath(name).name != name
                    or "\\" in name
                    or name == "manifest.json"
                    or not name.endswith(".json")
                    or name in names
                    or not isinstance(entry["complete"], bool)
                    or not isinstance(entry["parent"], dict)
                ):
                    raise ValueError("Invalid manifest entry")
                names.append(name)
                doc = files[name]
                resource = doc["value"]
                if (
                    doc["issue"]
                    or not isinstance(resource, dict)
                    or resource["runId"] != manifest["runId"]
                    or not isinstance(resource["complete"], bool)
                    or resource["parent"] != entry["parent"]
                    or resource["complete"] != entry["complete"]
                    or utc(resource["observedAt"]) > final_time
                ):
                    raise ValueError("Resource and manifest disagree")
                _validate_records(module, resource)
                expected_name, endpoint = _resource_name_endpoint(module, resource)
                if endpoint is not None and (
                    name != expected_name or resource["requestPath"] != endpoint
                ):
                    raise ValueError("Resource identity or endpoint mismatch")
                run["resources"].append(doc)
            if set(files) != {"manifest.json", *names}:
                raise ValueError("Unlisted or missing run resources")
            if module in PERMISSIONS:
                if len(entries) != 1:
                    raise ValueError("Permission slice must contain one resource")
            elif module == "fabricitemjobhistory" and (
                not 1 <= len(entries) <= 100
                or len({target(entry["parent"]) for entry in entries}) != len(entries)
            ):
                raise ValueError("Invalid job resource scope")
            elif manifest["complete"] != (
                all(entry["complete"] for entry in entries) and not manifest["errors"]
            ):
                raise ValueError("Manifest completeness disagrees with resources")
            run["valid"] = True
        except (ValueError, KeyError, TypeError, AttributeError, IndexError):
            run["issue"] = "invalid_run_artifacts"
    for path, run in result.items():
        if (
            run["module"] in PERMISSIONS
            and run["valid"]
            and run["manifest"]["finished"]
        ):
            try:
                run["chain"] = permission_chain(path, result)
            except (ValueError, KeyError, TypeError, AttributeError, IndexError):
                run["issue"] = "invalid_permission_chain"
                run["valid"] = False
    return result


def _integer(value):
    if type(value) is not int or value < 0:
        raise ValueError("Invalid nonnegative integer")
    return value


def permission_chain(path, runs):
    """Follow enumerated immutable manifests only, then validate cursor transitions."""
    module = runs[path]["module"]
    tenant = PERMISSIONS[module] == "Tenant"
    chain, seen = [], set()
    while True:
        if path in seen or len(seen) >= 10000 or path not in runs:
            raise ValueError("Missing, cyclic or excessive permission evidence")
        seen.add(path)
        run = runs[path]
        manifest = run["manifest"]
        if not run["valid"] or run["module"] != module or not manifest["finished"]:
            raise ValueError("Uncommitted permission evidence")
        chain.append(run)
        execution = manifest["execution"]
        links = execution["cycleEvidence"]
        if not isinstance(links, list) or len(links) > 1:
            raise ValueError("Invalid evidence links")
        if not links:
            break
        path = text(links[0])
    chain.reverse()
    selected, selected_keys, inventories, access = [], set(), set(), []
    before, healthy, prior, prior_parent, prior_task = 0, True, None, None, None
    task_keys, task_ids = set(), set()
    inventory_workspaces = set()
    for run in chain:
        manifest, resource = run["manifest"], run["resources"][0]["value"]
        e, parent = manifest["execution"], resource["parent"]
        inventory = parent.get("observationUnit") == "inventoryPage"
        for field in ("cycleFinished", "cycleHealthy", "deferred"):
            if not isinstance(e[field], bool):
                raise TypeError("Invalid cycle flag")
        attempts = _integer(e["attempts"])
        if attempts > 1:
            raise ValueError("Invalid slice attempt count")
        unattempted = e["deferred"] and attempts == 0
        if resource["complete"] and (attempts != 1 or e["deferred"]):
            raise ValueError("Successful deferred or unattempted resource")
        if prior is not None and (
            prior["execution"]["cycleFinished"]
            or utc(prior["observedAt"]) > utc(resource["observedAt"])
        ):
            raise ValueError("Invalid cycle transition")
        previous_healthy = healthy
        healthy = healthy and (resource["complete"] or unattempted)
        if e["cycleHealthy"] != healthy or (
            healthy and manifest["errors"] and not unattempted
        ):
            raise ValueError("Nonmonotonic cycle health")
        if tenant:
            if (
                e["scopeMode"] != "tenant"
                or e["phase"] != ("inventory" if inventory else "access")
                or not isinstance(e["inventoryComplete"], bool)
                or e["resourceOutcome"]
                not in {
                    "succeeded",
                    "failed",
                    "unsupported",
                    "denied",
                    "not_found",
                    "deferred",
                }
                or (e["resourceOutcome"] == "succeeded") != resource["complete"]
                or (e["resourceOutcome"] == "deferred") != e["deferred"]
                or (attempts == 0 and not e["deferred"])
            ):
                raise ValueError("Invalid tenant execution")
            if inventory:
                if before != 0 or (
                    prior is not None and prior["execution"]["inventoryComplete"]
                ):
                    raise ValueError("Inventory after tenant access")
                if resource["complete"]:
                    for item in resource["records"]:
                        key = target(item)[:2]
                        if key in selected_keys:
                            raise ValueError("Duplicate tenant inventory")
                        selected_keys.add(key)
                        selected.append(item)
                after = 0
                if e["inventoryComplete"] != parent["inventoryComplete"]:
                    raise ValueError("Inventory completion mismatch")
            else:
                if (
                    prior is None
                    or not prior["execution"]["inventoryComplete"]
                    or before >= len(selected)
                    or target(parent) != target(selected[before])
                    or not e["inventoryComplete"]
                ):
                    raise ValueError("Access outside tenant inventory order")
                after = before + (0 if unattempted else 1)
            if (
                _integer(e["discoveredItems"]) != len(selected)
                or _integer(e["nextItem"]) != after
                or e["cycleFinished"]
                != (e["inventoryComplete"] and after == len(selected))
            ):
                raise ValueError("Invalid tenant cursor")
        else:
            count = _integer(e["taskCount"])
            if not 2 <= count <= 100 or (
                prior is not None and count != prior["execution"]["taskCount"]
            ):
                raise ValueError("Invalid scoped task count")
            workspace = parent["workspaceId"]
            key = (workspace, "inventory") if inventory else target(parent)
            if prior is None:
                if not inventory:
                    raise ValueError("Cycle does not begin with inventory")
            elif before == prior_task:
                prior_key = (
                    (prior_parent["workspaceId"], "inventory")
                    if prior_parent.get("observationUnit") == "inventoryPage"
                    else target(prior_parent)
                )
                if key != prior_key:
                    raise ValueError("Deferred task changed identity")
            elif prior_parent.get("observationUnit") == "inventoryPage":
                if inventory or workspace != prior_parent["workspaceId"]:
                    raise ValueError("Invalid inventory to access transition")
            elif not inventory:
                if workspace != prior_parent["workspaceId"]:
                    raise ValueError("Access workspace changed without inventory")
            elif workspace in inventory_workspaces:
                raise ValueError("Repeated workspace inventory task")
            if inventory:
                inventory_workspaces.add(workspace)
                if resource["complete"]:
                    for item in resource["records"]:
                        identity = target(item)
                        if identity in selected_keys:
                            raise ValueError("Duplicate scoped inventory")
                        selected_keys.add(identity)
                        selected.append(item)
                    if parent["inventoryComplete"]:
                        inventories.add(workspace)
            elif resource["complete"]:
                if target(parent) not in selected_keys or workspace not in inventories:
                    raise ValueError("Access outside completed scoped inventory")
                access.append(target(parent))
            delta = (
                0
                if unattempted
                or (
                    inventory
                    and resource["complete"]
                    and not parent["inventoryComplete"]
                )
                else 1
            )
            after = before + delta
            if (
                _integer(e["nextTask"]) != after
                or after > count
                or e["cycleFinished"] != (after == count)
            ):
                raise ValueError("Invalid scoped cursor")
            task_keys.add((before, key))
            task_ids.add(before)
            if manifest["complete"] and (
                not 1 <= len(access) <= 50
                or len(access) != len(set(access))
                or set(access) != selected_keys
                or len(task_keys) != count
                or len(task_ids) != count
                or len({key[0] for key in access}) + len(access) != count
            ):
                raise ValueError("Incomplete scoped selection")
        if manifest["complete"] != (e["cycleFinished"] and healthy):
            raise ValueError("Invalid permission cycle completeness")
        # A failed slice cannot become healthy merely by following a newer link.
        if not previous_healthy and healthy:
            raise ValueError("Cycle health recovered without a new cycle")
        prior_task, before, prior, prior_parent = before, after, manifest, parent
    return {
        "nodes": chain,
        "cycle_id": chain[0]["manifest"]["runId"],
        "selected_items": len(selected),
    }


def _health(run):
    manifest = run["manifest"] or {}
    valid, finished = run["valid"], manifest.get("finished") is True
    resources = run["resources"]
    status = (
        "invalid"
        if not valid
        else "unfinished"
        if not finished
        else "success"
        if manifest["complete"]
        else "partial_or_failed"
    )
    return row(
        "collection_health",
        run["doc"],
        "run",
        coverage_status=status,
        complete=manifest.get("complete") if valid else None,
        finished=manifest.get("finished") if valid else None,
        successful_resources=sum(d["value"]["complete"] for d in resources)
        if valid and finished
        else None,
        failed_resources=sum(not d["value"]["complete"] for d in resources)
        if valid and finished
        else None,
        coverage=manifest.get("coverage")
        if isinstance(manifest.get("coverage"), str)
        else None,
        issue=run["issue"],
        cycle_id=run["chain"]["cycle_id"] if run["chain"] else None,
    )


def _permission_tables(runs, output):
    for run in runs.values():
        if run["module"] not in PERMISSIONS:
            continue
        manifest = run["manifest"] or {}
        chain = run["chain"]
        valid = run["valid"] and chain is not None
        cycle_id = (
            chain["cycle_id"]
            if valid
            else (
                manifest.get("runId")
                if isinstance(manifest.get("runId"), str)
                else None
            )
        )
        execution = manifest.get("execution") or {}
        output["permission_cycles"].append(
            row(
                "permission_cycles",
                run["doc"],
                "cycle_tip",
                cycle_id=cycle_id,
                chain_valid=valid,
                cycle_finished=execution.get("cycleFinished") if valid else None,
                cycle_healthy=execution.get("cycleHealthy") if valid else None,
                slice_count=len(chain["nodes"]) if valid else None,
                selected_items=chain["selected_items"] if valid else None,
                coverage_status="success"
                if valid and manifest["complete"]
                else "partial_or_failed"
                if valid
                else "invalid",
            )
        )
        if not valid:
            continue
        doc = run["resources"][0]
        resource, parent = doc["value"], doc["value"]["parent"]
        inventory = parent.get("observationUnit") == "inventoryPage"
        fields = (
            {"workspace_id": parent.get("workspaceId")}
            if inventory
            else target_fields(parent)
        )
        successful = resource["complete"]
        output["permission_resources"].append(
            row(
                "permission_resources",
                doc,
                "resource",
                **fields,
                cycle_id=cycle_id,
                observation_unit="inventory_page" if inventory else "item_access",
                record_count=len(resource["records"]) if successful else None,
                resource_outcome=execution.get("resourceOutcome")
                or (
                    "succeeded"
                    if successful
                    else "deferred"
                    if execution["deferred"]
                    else "failed"
                ),
                coverage_status="success"
                if successful
                else "deferred"
                if execution["deferred"]
                else "failed",
            )
        )
        if not successful:
            continue
        for record in resource["records"]:
            if inventory:
                output["inventory"].append(
                    row(
                        "inventory",
                        doc,
                        target(record),
                        **target_fields(record),
                        inventory_source="permission_inventory",
                        cycle_id=cycle_id,
                    )
                )
            else:
                principal, detail = record["principal"], record["itemAccessDetails"]
                permissions = sorted(set(detail["permissions"]))
                additional = detail.get("additionalPermissions")
                additional = sorted(set(additional)) if additional is not None else None
                output["permission_grants"].append(
                    row(
                        "permission_grants",
                        doc,
                        [principal["id"], principal["type"], permissions, additional],
                        **fields,
                        cycle_id=cycle_id,
                        principal_id=principal["id"],
                        principal_type=principal["type"],
                        permissions_json=canonical(permissions),
                        additional_permissions_json=canonical(additional),
                    )
                )


def _job_tables(runs, output):
    for run in runs.values():
        if run["module"] != "fabricitemjobhistory" or not run["valid"]:
            continue
        if not run["manifest"]["finished"]:
            continue
        for doc in run["resources"]:
            resource, parent = doc["value"], doc["value"]["parent"]
            fields = target_fields(parent)
            output["job_resources"].append(
                row(
                    "job_resources",
                    doc,
                    "resource",
                    **fields,
                    coverage_status="success" if resource["complete"] else "failed",
                    record_count=len(resource["records"])
                    if resource["complete"]
                    else None,
                )
            )
            if not resource["complete"]:
                continue
            for job in resource["records"]:
                start = utc(job["startTimeUtc"]) if job.get("startTimeUtc") else None
                end = utc(job["endTimeUtc"]) if job.get("endTimeUtc") else None
                duration = (
                    (
                        datetime.fromisoformat(end.replace("Z", "+00:00"))
                        - datetime.fromisoformat(start.replace("Z", "+00:00"))
                    ).total_seconds()
                    if start and end
                    else None
                )
                failure = job.get("failureReason") or {}
                output["job_observations"].append(
                    row(
                        "job_observations",
                        doc,
                        job["id"],
                        **fields,
                        job_id=job["id"],
                        job_type=job["jobType"],
                        invoke_type=job["invokeType"],
                        status=job["status"],
                        start_time_utc=start,
                        end_time_utc=end,
                        duration_seconds=duration
                        if duration is not None and duration >= 0
                        else None,
                        root_activity_id=job.get("rootActivityId"),
                        failure_code=failure.get("errorCode")
                        if isinstance(failure.get("errorCode"), str)
                        else None,
                    )
                )


def _legacy_inventory(doc, output):
    path = doc["path"]
    module = path.split("/")[0]
    try:
        if not re.fullmatch(r"(items|workspaces)/\d{4}/\d{2}/\d{2}/[^/]+\.json", path):
            raise ValueError("Invalid legacy inventory path")
        values = records(doc["value"])
        datetime.strptime("/".join(path.split("/")[1:4]), "%Y/%m/%d")
        pending = []
        for value in values:
            item = uuid(value["id"])
            workspace = item if module == "workspaces" else uuid(value["workspaceId"])
            kind = "Workspace" if module == "workspaces" else text(value["type"])
            name = value.get("displayName", value.get("name"))
            pending.append(
                row(
                    "inventory",
                    doc,
                    [workspace, item, kind],
                    workspace_id=workspace,
                    item_id=item,
                    item_type=kind,
                    item_name=name if isinstance(name, str) else None,
                    inventory_source="legacy_" + module,
                    coverage_status="legacy_unverified",
                )
            )
        output["inventory"].extend(pending)
        status, count = "legacy_unverified", len(values)
    except (ValueError, KeyError, TypeError, AttributeError):
        status, count = "invalid", None
    output["collection_health"].append(
        row(
            "collection_health",
            doc,
            "legacy_file",
            coverage_status=status,
            record_count=count,
            issue="no_manifest_or_exact_observation_time",
        )
    )


def _scanner_tables(doc, output, *, legacy_snapshot=None):
    try:
        legacy = legacy_snapshot is not None
        scan = legacy_snapshot if legacy else doc["value"]
        if (
            doc["issue"]
            or (
                not legacy
                and not re.fullmatch(
                    r"lineage_edges/\d{4}/\d{2}/\d{2}/[A-Za-z0-9][A-Za-z0-9_-]{0,127}\.lineage_edges\.json",
                    doc["path"],
                )
            )
            or type(scan["schemaVersion"]) is not int
            or scan["schemaVersion"] != 1
            or scan["source"] != "Scanner"
            or scan["complete"] is not True
            or (
                not legacy
                and PurePosixPath(doc["path"]).name
                != scan["scanId"] + ".lineage_edges.json"
            )
            or scan["scanMode"] not in {"full", "incremental"}
        ):
            raise ValueError("Invalid scanner envelope")
        observed = None if legacy else utc(scan["observedAtUtc"])
        status = "legacy_unverified" if legacy else "success"
        date_parts = (
            doc["path"].split("/")[2:5] if legacy else doc["path"].split("/")[1:4]
        )
        datetime.strptime("/".join(date_parts), "%Y/%m/%d")
        requested = {uuid(value) for value in scan["requestedWorkspaceIds"]}
        returned = {uuid(value) for value in scan["returnedWorkspaceIds"]}
        if not returned.issubset(requested):
            raise ValueError("Unexpected returned workspace")
        items = records(scan["items"])
        indexed = {}
        inventory = []
        for item in items:
            identity, workspace = uuid(item["id"]), uuid(item["workspaceId"])
            if identity in indexed or workspace not in returned:
                raise ValueError("Invalid scanner item inventory")
            indexed[identity] = item
            inventory.append(
                row(
                    "inventory",
                    doc,
                    identity,
                    observed=observed,
                    workspace_id=workspace,
                    item_id=identity,
                    item_type=text(item["type"]),
                    item_name=item.get("name")
                    if isinstance(item.get("name"), str)
                    else None,
                    inventory_source="legacy_scanner" if legacy else "scanner_batch",
                    coverage_status=status,
                )
            )
        edges = []
        for edge in records(scan["edges"]):
            item = indexed[edge["QueryItemId"]]
            if (
                edge["Source"] != "Scanner"
                or edge["Direction"] not in {"upstream", "downstream"}
                or edge["QueryWorkspaceId"] != item["workspaceId"]
                or edge["QueryItemType"] != item["type"]
            ):
                raise ValueError("Invalid scanner edge scope")
            related = text(edge["RelatedItemId"])
            if edge.get("RelatedItemType") is not None:
                text(edge["RelatedItemType"])
            if edge.get("RelatedWorkspaceId") is not None:
                uuid(edge["RelatedWorkspaceId"])
            key = [item["id"], edge["Direction"], related, text(edge["RelationType"])]
            edges.append(
                row(
                    "lineage_edges",
                    doc,
                    key,
                    observed=observed,
                    workspace_id=item["workspaceId"],
                    item_id=item["id"],
                    item_type=item["type"],
                    related_workspace_id=edge.get("RelatedWorkspaceId"),
                    related_item_id=related,
                    related_type=edge.get("RelatedItemType"),
                    direction=edge["Direction"],
                    relation_type=edge["RelationType"],
                    evidence_type="scanner_relationship",
                    runtime_verified=False,
                    coverage_status=status,
                    resolution="in_batch"
                    if related in indexed
                    else "external_or_unresolved",
                )
            )
        coverage = status if requested == returned else "partial_batch"
        resources = [
            row(
                "lineage_resources",
                doc,
                ["scanner_item", identity],
                observed=observed,
                workspace_id=item["workspaceId"],
                item_id=identity,
                item_type=item["type"],
                evidence_type="scanner_relationship",
                record_count=sum(edge["item_id"] == identity for edge in edges),
                diagnostic_count=0,
                coverage_status=coverage,
            )
            for identity, item in indexed.items()
        ]
        output["inventory"].extend(inventory)
        output["lineage_edges"].extend(edges)
        output["lineage_resources"].extend(resources)
        output["collection_health"].append(
            row(
                "collection_health",
                doc,
                "scanner_batch",
                observed=observed,
                complete=None if legacy else requested == returned,
                finished=None if legacy else True,
                record_count=len(items),
                coverage_status=coverage,
                coverage="Scanner batch only; not exhaustive tenant or runtime I/O coverage",
                issue="no_manifest_or_exact_observation_time" if legacy else None,
            )
        )
    except (ValueError, KeyError, TypeError, AttributeError):
        output["collection_health"].append(
            row(
                "collection_health",
                doc,
                "scanner_batch",
                coverage_status="invalid",
                issue="invalid_scanner_envelope",
            )
        )


def _definition_tables(runs, output):
    for run in runs.values():
        if (
            run["module"] != "fabricitemdefinitions"
            or not run["valid"]
            or not run["manifest"]["finished"]
        ):
            continue
        for doc in run["resources"]:
            resource, parent = doc["value"], doc["value"]["parent"]
            if not parent:
                continue
            fields = target_fields(parent)
            status, edges, diagnostics = "failed", [], []
            if resource["complete"]:
                try:
                    values = records(resource["records"])
                    if len(values) != 1:
                        raise ValueError("Invalid definition record count")
                    key = (
                        "notebookReferences"
                        if parent["type"] == "Notebook"
                        else "pipelineReferences"
                    )
                    refs = values[0][key]
                    if (
                        type(refs["schemaVersion"]) is not int
                        or refs["schemaVersion"] != 1
                        or refs["evidence"] != "static_inference"
                        or refs["runtimeVerified"] is not False
                    ):
                        raise ValueError("Invalid static evidence")
                    for ref in records(refs["references"]):
                        if ref["operation"] not in {"read", "write", "execute"}:
                            raise ValueError("Invalid static operation")
                        reference = text(ref["reference"])
                        kind = text(ref["kind"])
                        if ref.get("workspaceId") is not None:
                            uuid(ref["workspaceId"])
                        edges.append(
                            row(
                                "lineage_edges",
                                doc,
                                ref,
                                **fields,
                                direction="downstream"
                                if ref["operation"] == "write"
                                else "upstream",
                                related_workspace_id=ref.get("workspaceId"),
                                related_item_id=reference
                                if ref.get("referenceType") == "id"
                                else None,
                                related_type=kind,
                                relation_type=ref["operation"],
                                reference=reference,
                                resolution="unresolved",
                                evidence_type="static_inference",
                                runtime_verified=False,
                            )
                        )
                    for index, diagnostic in enumerate(records(refs["diagnostics"])):
                        diagnostics.append(
                            row(
                                "lineage_diagnostics",
                                doc,
                                index,
                                **fields,
                                reason=text(diagnostic["reason"]),
                                location=canonical(
                                    {
                                        key: diagnostic[key]
                                        for key in ("cell", "line", "fieldPath")
                                        if key in diagnostic
                                    }
                                ),
                            )
                        )
                    status = "success"
                except (ValueError, KeyError, TypeError, AttributeError):
                    status, edges, diagnostics = "invalid", [], []
            output["lineage_resources"].append(
                row(
                    "lineage_resources",
                    doc,
                    "definition",
                    **fields,
                    coverage_status=status,
                    evidence_type="static_inference",
                    record_count=len(edges) if status == "success" else None,
                    diagnostic_count=len(diagnostics) if status == "success" else None,
                )
            )
            output["lineage_edges"].extend(edges)
            output["lineage_diagnostics"].extend(diagnostics)


def transform(
    documents, *, scanner_extractor=None, estate_extractor=None, module_extractor=None
):
    """Return historical observations and explicit coverage, never inferred deletions."""
    documents = list(documents)
    if len({doc["path"] for doc in documents}) != len(documents):
        raise ValueError("Duplicate source path")
    output = {table: [] for table in SCHEMAS}
    runs = build_runs(documents)
    output["collection_health"].extend(_health(run) for run in runs.values())
    _permission_tables(runs, output)
    _job_tables(runs, output)
    _definition_tables(runs, output)
    present = {doc["path"].split("/")[0] for doc in documents}
    for doc in documents:
        module = doc["path"].split("/")[0]
        if module in {"items", "workspaces"}:
            _legacy_inventory(doc, output)
        elif module == "lineage_edges":
            _scanner_tables(doc, output)
        elif module == "catalog" and doc["path"].startswith("catalog/scans/"):
            try:
                if scanner_extractor is None or not re.fullmatch(
                    r"catalog/scans/\d{4}/\d{2}/\d{2}/\d{8}_\d+\.scanResults(?:\.fullscan)?\.json",
                    doc["path"],
                ):
                    raise ValueError("Unsupported raw scanner artifact")
                scan = doc["value"]
                workspaces = [
                    workspace["id"] for workspace in records(scan["workspaces"])
                ]
                snapshot = scanner_extractor(
                    scan, "legacy_" + doc["sha256"][:32], workspaces, None
                )
                _scanner_tables(doc, output, legacy_snapshot=snapshot)
            except (ValueError, KeyError, TypeError, AttributeError):
                output["collection_health"].append(
                    row(
                        "collection_health",
                        doc,
                        "legacy_scanner",
                        coverage_status="invalid",
                        issue="invalid_raw_scanner",
                    )
                )
    for module in (*MODULES, "lineage_edges", "items", "workspaces"):
        if module not in present:
            output["collection_health"].append(
                row(
                    "collection_health",
                    {"path": module + "/_missing", "sha256": "", "value": None},
                    "missing",
                    coverage_status="missing",
                    issue="no_collection_files",
                )
            )
    if estate_extractor is not None:
        estate_extractor(documents, runs, output)
    if module_extractor is not None:
        module_extractor(documents, runs, output)
    for table, rows in output.items():
        unique = {}
        for entry in rows:
            entry["record_sha256"] = sha256(
                canonical(
                    {
                        key: value
                        for key, value in entry.items()
                        if key != "record_sha256"
                    }
                ).encode()
            )
            existing = unique.get(entry["observation_id"])
            if existing is not None and existing != entry:
                raise ValueError("Conflicting observation identity")
            unique[entry["observation_id"]] = entry
        output[table] = sorted(
            unique.values(), key=lambda entry: entry["observation_id"]
        )
    return output


def latest(rows, keys):
    """Select latest observations inside an explicitly chosen scope."""
    selected = {}
    for entry in rows:
        key = tuple(entry[field] for field in keys)
        order = (entry["observed_at_utc"] or "", entry["source_path"])
        if key not in selected or order > (
            selected[key]["observed_at_utc"] or "",
            selected[key]["source_path"],
        ):
            selected[key] = entry
    return list(selected.values())


def selected_permission_cycle(tables, mode):
    if mode not in {"Scoped", "Tenant"}:
        raise ValueError("Select Scoped or Tenant explicitly")
    candidates = [
        entry for entry in tables["permission_cycles"] if entry["mode"] == mode
    ]
    if not candidates:
        return None
    unknown = [entry for entry in candidates if entry["observed_at_utc"] is None]
    if unknown:
        return max(unknown, key=lambda entry: entry["source_path"])
    return max(
        candidates,
        key=lambda entry: (entry["observed_at_utc"] or "", entry["source_path"]),
    )


def table_name(prefix, dataset):
    if not re.fullmatch(r"monitor_[A-Za-z0-9_]*", prefix) or dataset not in SCHEMAS:
        raise ValueError("Use a monitor_ table prefix and a known dataset")
    return prefix + dataset


def to_frame(spark_session, dataset, rows):
    from pyspark.sql import types as T

    types = {
        "int": T.IntegerType,
        "long": T.LongType,
        "double": T.DoubleType,
        "boolean": T.BooleanType,
        "string": T.StringType,
    }
    schema = T.StructType(
        [
            T.StructField(name, types[kind](), True)
            for name, kind in {**COMMON, **SCHEMAS[dataset]}.items()
        ]
    )
    return spark_session.createDataFrame(rows, schema=schema)


def publish(spark_session, tables, datasets, prefix, *, write=False):
    """Insert immutable observations only; never overwrite, delete or run VACUUM."""
    from delta.tables import DeltaTable
    from pyspark.sql import functions as F

    spark_session.conf.set("spark.sql.session.timeZone", "UTC")
    prepared = {}
    for dataset in datasets:
        name = table_name(prefix, dataset)
        frame = to_frame(spark_session, dataset, tables[dataset])
        exists = spark_session.catalog.tableExists(name)
        if exists:
            properties = {
                entry["key"]: entry["value"]
                for entry in spark_session.sql(f"SHOW TBLPROPERTIES {name}").collect()
            }
            if (
                properties.get("monitor.contract") != str(SCHEMA_VERSION)
                or properties.get("monitor.dataset") != dataset
            ):
                raise ValueError("Existing table is not owned by this Monitor contract")
            existing = DeltaTable.forName(spark_session, name).toDF()
            expected_types = {
                field.name: field.dataType for field in frame.schema.fields
            }
            if {
                field.name: field.dataType for field in existing.schema.fields
            } != expected_types:
                raise ValueError("Existing table schema differs; select a new prefix")
            conflict = (
                existing.alias("old")
                .join(frame.alias("new"), "observation_id")
                .filter(
                    ~F.col("old.record_sha256").eqNullSafe(F.col("new.record_sha256"))
                )
            )
            if conflict.limit(1).count():
                raise ValueError(
                    "Immutable observation changed; no overwrite permitted"
                )
            changed_source = (
                existing.alias("old")
                .join(frame.alias("new"), "source_path")
                .filter(
                    ~F.col("old.source_sha256").eqNullSafe(F.col("new.source_sha256"))
                    & ~F.col("old.coverage_status").isin(
                        "invalid", "unfinished", "missing", "legacy_unverified"
                    )
                )
            )
            if changed_source.limit(1).count():
                raise ValueError("Committed source changed; no new version permitted")
        prepared[dataset] = (name, frame, exists)
    if not write:
        print(
            "PREVIEW ONLY: no Delta tables written. Set WRITE_TABLES=True to persist."
        )
        return {dataset: len(tables[dataset]) for dataset in datasets}
    for dataset, (name, frame, exists) in prepared.items():
        if not exists:
            columns = ", ".join(
                f"{column} {kind}"
                for column, kind in {**COMMON, **SCHEMAS[dataset]}.items()
            )
            spark_session.sql(
                f"CREATE TABLE {name} ({columns}) USING DELTA "
                f"TBLPROPERTIES ('monitor.contract'='{SCHEMA_VERSION}', "
                f"'monitor.dataset'='{dataset}')"
            )
        if tables[dataset]:
            DeltaTable.forName(spark_session, name).alias("old").merge(
                frame.alias("new"), "old.observation_id = new.observation_id"
            ).whenNotMatchedInsertAll().execute()
    print("Published immutable observations; no cross-table transaction is promised.")
    return {dataset: len(tables[dataset]) for dataset in datasets}


def read_curated(spark_session, prefix, dataset):
    name = table_name(prefix, dataset)
    if not spark_session.catalog.tableExists(name):
        raise ValueError(
            f"Missing {name}; run its transformation with WRITE_TABLES=True"
        )
    properties = {
        entry["key"]: entry["value"]
        for entry in spark_session.sql(f"SHOW TBLPROPERTIES {name}").collect()
    }
    if (
        properties.get("monitor.contract") != str(SCHEMA_VERSION)
        or properties.get("monitor.dataset") != dataset
    ):
        raise ValueError("Selected table does not match the Monitor contract")
    return spark_session.table(name)


def latest_frame(frame, keys):
    from pyspark.sql import Window
    from pyspark.sql import functions as F

    order = Window.partitionBy(*keys).orderBy(
        F.col("observed_at_utc").desc_nulls_last(), F.col("source_path").desc()
    )
    return (
        frame.withColumn("_rank", F.row_number().over(order))
        .filter(F.col("_rank") == 1)
        .drop("_rank")
    )


def plot_counts(frame, category, title, *, limit=20):
    """Collect only a bounded aggregate, never arbitrary raw principal/job rows."""
    import matplotlib.pyplot as plt
    from pyspark.sql import functions as F

    values = (
        frame.groupBy(category)
        .count()
        .orderBy(F.desc("count"), F.asc(category))
        .limit(limit)
        .toPandas()
    )
    if values.empty:
        print(title + ": no matching observations; inspect the coverage tables.")
        return
    values[category] = values[category].fillna("unknown")
    axis = values.plot.bar(x=category, y="count", legend=False, title=title)
    axis.set_ylabel("Observations")
    plt.tight_layout()
    plt.show()


def preview_inventory(spark_session, display_frame, tables):
    display_frame(
        to_frame(spark_session, "collection_health", tables["collection_health"])
        .groupBy("module", "coverage_status")
        .count()
        .orderBy("module", "coverage_status")
    )
    display_frame(
        to_frame(spark_session, "inventory", tables["inventory"])
        .groupBy("inventory_source", "item_type", "coverage_status")
        .count()
    )


def preview_permissions(spark_session, display_frame, tables, mode):
    selected = selected_permission_cycle(tables, mode)
    if selected is None:
        print("No selected-mode cycle. Missing evidence is not a zero-grant success.")
        return
    print(
        {
            key: selected[key]
            for key in (
                "mode",
                "cycle_id",
                "coverage_status",
                "chain_valid",
                "cycle_finished",
                "cycle_healthy",
                "selected_items",
            )
        }
    )
    resources = [
        entry
        for entry in tables["permission_resources"]
        if entry["mode"] == mode and entry["cycle_id"] == selected["cycle_id"]
    ]
    display_frame(
        to_frame(spark_session, "permission_resources", resources)
        .groupBy("observation_unit", "coverage_status")
        .count()
    )


def preview_jobs(spark_session, display_frame, tables):
    current_jobs = latest(
        tables["job_observations"], ["workspace_id", "item_id", "job_id"]
    )
    display_frame(
        to_frame(spark_session, "job_resources", tables["job_resources"])
        .groupBy("coverage_status")
        .count()
    )
    display_frame(
        to_frame(spark_session, "job_observations", current_jobs)
        .groupBy("item_type", "status")
        .count()
    )


def preview_lineage(spark_session, display_frame, tables):
    for dataset, keys in (
        ("lineage_resources", ["evidence_type", "coverage_status"]),
        ("lineage_edges", ["evidence_type", "direction", "related_type"]),
        ("lineage_diagnostics", ["reason"]),
    ):
        display_frame(
            to_frame(spark_session, dataset, tables[dataset]).groupBy(*keys).count()
        )


ESTATE_DOMAINS = {
    "activity": ("activity_events",),
    "catalog": (
        "catalog_workspaces",
        "catalog_items",
        "catalog_models",
        "model_tables",
        "model_columns",
        "model_measures",
        "model_relationships",
    ),
    "connections": ("connection_observations",),
    "gateways": ("gateway_observations", "gateway_datasources", "gateway_status"),
    "refresh": ("refresh_observations", "refreshable_observations"),
}
ESTATE_SOURCES = {
    "activity": ("activity",),
    "catalog": ("catalog", "items", "workspaces"),
    "connections": ("fabricconnections",),
    "gateways": ("gateways",),
    "refresh": ("datasetrefresh", "datasetrefreshable"),
}
SCANNER_KINDS = {
    "reports": "Report",
    "datasets": "SemanticModel",
    "dataflows": "Dataflow",
    "dashboards": "Dashboard",
    "datamarts": "Datamart",
}


def estate_optional(value, kind=str):
    if value is not None and type(value) is not kind:
        raise TypeError("Invalid optional metadata")
    return value


def estate_utc(value):
    if value is None:
        return None
    parsed = datetime.fromisoformat(text(value).replace("Z", "+00:00"))
    # Activity CreationTime and dataset refresh times are provider-defined UTC.
    return utc(parsed.replace(tzinfo=parsed.tzinfo or timezone.utc).isoformat())


def estate_source_date(doc):
    parts = PurePosixPath(doc["path"]).parts
    offset = 2 if parts[:2] == ("catalog", "scans") else 1
    return datetime.strptime("/".join(parts[offset : offset + 3]), "%Y/%m/%d").strftime(
        "%Y-%m-%d"
    )


def estate_records(doc):
    if doc["issue"]:
        raise ValueError("Malformed raw JSON")
    value = doc["value"]
    if isinstance(value, str):
        value = json.loads(value)
    if isinstance(value, dict):
        value = value["value"]
    return records(value)


def estate_optional_records(value):
    return records(value) if value is not None else None


def estate_row(dataset, doc, key, **values):
    if "source_date" not in values:
        values["source_date"] = estate_source_date(doc)
    result = row(
        dataset,
        doc,
        key,
        coverage_status=values.pop("coverage_status", "legacy_unverified"),
        **values,
    )
    result["observed_at_utc"], result["run_id"] = None, None
    return result


def estate_activity(doc, pending):
    if not re.fullmatch(r"activity/\d{4}/\d{2}/\d{2}/\d{8}_\d+\.json", doc["path"]):
        raise ValueError("Unsupported activity file")
    for index, event in enumerate(estate_records(doc)):
        identity = estate_optional(event.get("Id"))
        if identity is not None:
            text(identity)
        actor = estate_optional(event.get("UserId") or event.get("UserKey"))
        actor_type = event.get("UserType")
        if actor_type is not None and type(actor_type) not in (int, str):
            raise TypeError("Invalid actor type")
        pending["activity_events"].append(
            estate_row(
                "activity_events",
                doc,
                identity or index,
                event_id=identity,
                event_time_utc=estate_utc(event.get("CreationTime")),
                operation=text(event["Operation"]),
                workload=estate_optional(event.get("Workload")),
                actor_key=sha256(actor.encode()) if actor else None,
                actor_type=str(actor_type) if actor_type is not None else None,
                is_success=estate_optional(event.get("IsSuccess"), bool),
                workspace_id=estate_optional(
                    event.get("WorkspaceId") or event.get("WorkSpaceId")
                ),
                item_id=estate_optional(
                    event.get("ArtifactId")
                    or event.get("DatasetId")
                    or event.get("ReportId")
                ),
                item_type=estate_optional(event.get("ItemType")),
            )
        )


def estate_catalog(doc, pending):
    if doc["issue"] or not re.fullmatch(
        r"catalog/scans/\d{4}/\d{2}/\d{2}/\d{8}_\d+\.scanResults(?:\.fullscan)?\.json",
        doc["path"],
    ):
        raise ValueError("Unsupported catalog file")
    for workspace in records(doc["value"]["workspaces"]):
        workspace_id = text(workspace["id"])
        returned = [name for name in SCANNER_KINDS if workspace.get(name) is not None]
        pending["catalog_workspaces"].append(
            estate_row(
                "catalog_workspaces",
                doc,
                workspace_id,
                workspace_id=workspace_id,
                workspace_name=estate_optional(workspace.get("name")),
                workspace_type=estate_optional(workspace.get("type")),
                state=estate_optional(workspace.get("state")),
                capacity_id=estate_optional(workspace.get("capacityId")),
                returned_collections_json=canonical(returned),
            )
        )
        for collection, kind in SCANNER_KINDS.items():
            for item in estate_optional_records(workspace.get(collection)) or []:
                fields = {
                    "workspace_id": workspace_id,
                    "item_id": text(item["id"]),
                    "item_type": kind,
                }
                identity = [workspace_id, kind, fields["item_id"]]
                pending["catalog_items"].append(
                    estate_row(
                        "catalog_items",
                        doc,
                        identity,
                        **fields,
                        item_name=estate_optional(item.get("name")),
                    )
                )
                if kind != "SemanticModel":
                    continue
                tables = estate_optional_records(item.get("tables"))
                relationships = estate_optional_records(item.get("relationships"))
                pending["catalog_models"].append(
                    estate_row(
                        "catalog_models",
                        doc,
                        identity,
                        **fields,
                        model_name=estate_optional(item.get("name")),
                        schema_coverage="returned"
                        if tables is not None
                        else "not_returned",
                        table_count=len(tables) if tables is not None else None,
                        relationship_count=len(relationships)
                        if relationships is not None
                        else None,
                    )
                )
                for table in tables or []:
                    table_name = text(table["name"])
                    columns = estate_optional_records(table.get("columns"))
                    measures = estate_optional_records(table.get("measures"))
                    pending["model_tables"].append(
                        estate_row(
                            "model_tables",
                            doc,
                            [*identity, table_name],
                            **fields,
                            table_name=table_name,
                            is_hidden=estate_optional(table.get("isHidden"), bool),
                            column_count=len(columns) if columns is not None else None,
                            measure_count=len(measures)
                            if measures is not None
                            else None,
                        )
                    )
                    for column in columns or []:
                        name = text(column["name"])
                        pending["model_columns"].append(
                            estate_row(
                                "model_columns",
                                doc,
                                [*identity, table_name, name],
                                **fields,
                                table_name=table_name,
                                column_name=name,
                                data_type=estate_optional(column.get("dataType")),
                                is_hidden=estate_optional(column.get("isHidden"), bool),
                            )
                        )
                    for measure in measures or []:
                        name = text(measure["name"])
                        expression = estate_optional(measure.get("expression"))
                        pending["model_measures"].append(
                            estate_row(
                                "model_measures",
                                doc,
                                [*identity, table_name, name],
                                **fields,
                                table_name=table_name,
                                measure_name=name,
                                data_type=estate_optional(measure.get("dataType")),
                                expression_returned=bool(expression)
                                if "expression" in measure
                                else None,
                            )
                        )
                for index, relationship in enumerate(relationships or []):
                    pending["model_relationships"].append(
                        estate_row(
                            "model_relationships",
                            doc,
                            [*identity, "relationship", index],
                            **fields,
                            from_table=text(relationship["fromTable"]),
                            from_column=text(relationship["fromColumn"]),
                            to_table=text(relationship["toTable"]),
                            to_column=text(relationship["toColumn"]),
                            from_cardinality=estate_optional(
                                relationship.get("fromCardinality")
                            ),
                            to_cardinality=estate_optional(
                                relationship.get("toCardinality")
                            ),
                            is_active=estate_optional(
                                relationship.get("isActive"), bool
                            ),
                        )
                    )


def estate_gateways(doc, pending):
    name = PurePosixPath(doc["path"]).name
    if not re.fullmatch(r"gateways/\d{4}/\d{2}/\d{2}/[^/]+\.json", doc["path"]):
        raise ValueError("Unsupported gateway file")
    for entry in estate_records(doc):
        if name == "gateways.json":
            pending["gateway_observations"].append(
                estate_row(
                    "gateway_observations",
                    doc,
                    text(entry["id"]),
                    gateway_id=entry["id"],
                    gateway_name=estate_optional(entry.get("name")),
                    gateway_type=estate_optional(entry.get("type")),
                )
            )
        elif name in {"datasources.json", "datasource_details.json"}:
            gateway, datasource = text(entry["gatewayId"]), text(entry["id"])
            pending["gateway_datasources"].append(
                estate_row(
                    "gateway_datasources",
                    doc,
                    [gateway, datasource],
                    gateway_id=gateway,
                    datasource_id=datasource,
                    datasource_name=estate_optional(entry.get("datasourceName")),
                    datasource_type=estate_optional(entry.get("datasourceType")),
                    detail_returned=name == "datasource_details.json",
                )
            )
        elif name == "status.json":
            gateway, datasource = text(entry["gatewayId"]), text(entry["datasourceId"])
            error = entry.get("error")
            if not isinstance(error, dict):
                raise TypeError("Invalid gateway status")
            code = estate_optional(error.get("code"))
            detail = error.get("pbi.error", {})
            if not isinstance(detail, dict):
                raise TypeError("Invalid gateway status details")
            http = next(
                (
                    value["detail"]
                    for value in records(detail.get("details", []))
                    if type(value.get("detail")) is int
                    and 100 <= value["detail"] <= 599
                ),
                None,
            )
            pending["gateway_status"].append(
                estate_row(
                    "gateway_status",
                    doc,
                    [gateway, datasource],
                    gateway_id=gateway,
                    datasource_id=datasource,
                    status_code=code,
                    http_status=http,
                    connectivity_observation="success_response"
                    if code == "success"
                    else "failed_or_unknown",
                )
            )
        else:
            raise ValueError("Unsupported gateway artifact")


def estate_count(value):
    if value is not None and (type(value) is not int or value < 0):
        raise ValueError("Invalid optional count")
    return value


def estate_refresh(doc, pending):
    history = doc["path"].split("/")[0] == "datasetrefresh"
    expected = (
        "workspaces.datasets.refreshes.json"
        if history
        else "workspaces.datasets.refreshable.json"
    )
    if PurePosixPath(doc["path"]).name != expected:
        raise ValueError("Unsupported refresh file")
    for index, entry in enumerate(estate_records(doc)):
        if history:
            fields = {
                "workspace_id": text(entry["workspace_id"]),
                "item_id": text(entry["dataset_id"]),
                "item_type": "SemanticModel",
            }
            identity = entry.get("requestId", entry.get("id"))
            if identity is not None and type(identity) not in (int, str):
                raise TypeError("Invalid refresh identity")
            if isinstance(identity, str):
                text(identity)
            start, end = (
                estate_utc(entry.get("startTime")),
                estate_utc(entry.get("endTime")),
            )
            duration = None
            if start and end:
                delta = (
                    datetime.fromisoformat(end.replace("Z", "+00:00"))
                    - datetime.fromisoformat(start.replace("Z", "+00:00"))
                ).total_seconds()
                duration = delta if delta >= 0 else None
            failure_code = None
            failure = entry.get("serviceExceptionJson")
            if isinstance(failure, str):
                try:
                    parsed = json.loads(failure)
                    if isinstance(parsed, dict):
                        failure_code = estate_optional(
                            parsed.get("errorCode") or parsed.get("code")
                        )
                except (ValueError, TypeError):
                    failure_code = None
            pending["refresh_observations"].append(
                estate_row(
                    "refresh_observations",
                    doc,
                    [
                        fields["workspace_id"],
                        fields["item_id"],
                        identity if identity is not None else index,
                    ],
                    **fields,
                    dataset_name=estate_optional(entry.get("dataset_name")),
                    refresh_id=str(identity) if identity is not None else None,
                    refresh_type=estate_optional(entry.get("refreshType")),
                    status=estate_optional(entry.get("status")),
                    start_time_utc=start,
                    end_time_utc=end,
                    duration_seconds=duration,
                    failure_code=failure_code,
                )
            )
        else:
            group, capacity, schedule = (
                {} if entry.get(key) is None else entry[key]
                for key in ("group", "capacity", "refreshSchedule")
            )
            if not isinstance(group, dict) or not isinstance(capacity, dict):
                raise TypeError("Invalid refreshable parent")
            if not isinstance(schedule, dict):
                raise TypeError("Invalid refresh schedule")
            identity = text(entry["id"])
            kind = estate_optional(entry.get("kind"))
            pending["refreshable_observations"].append(
                estate_row(
                    "refreshable_observations",
                    doc,
                    identity,
                    workspace_id=estate_optional(group.get("id")),
                    item_id=identity,
                    item_type="SemanticModel"
                    if kind in {"Dataset", "SemanticModel"}
                    else kind or "Unknown",
                    dataset_name=estate_optional(entry.get("name")),
                    capacity_id=estate_optional(capacity.get("id")),
                    refresh_count=estate_count(entry.get("refreshCount")),
                    refresh_failures=estate_count(entry.get("refreshFailures")),
                    schedule_enabled=estate_optional(schedule.get("enabled"), bool),
                )
            )


def estate_connections(runs, output):
    for run in runs.values():
        if run["module"] != "fabricconnections":
            continue
        manifest = run["manifest"]
        status, issue, pending = "invalid", "invalid_connection_run", []
        if run["valid"] and not manifest["finished"]:
            status, issue = "unfinished", "unfinished_connection_run"
        elif run["valid"]:
            try:
                if len(run["resources"]) != 1:
                    raise ValueError("Invalid connection resource count")
                doc = run["resources"][0]
                resource = doc["value"]
                if (
                    PurePosixPath(doc["path"]).name != "connections.json"
                    or resource["requestPath"] != "connections"
                    or resource["parent"]
                ):
                    raise ValueError("Invalid connection resource identity")
                if resource["complete"]:
                    for entry in records(resource["records"]):
                        details = entry["connectionDetails"]
                        if not isinstance(details, dict):
                            raise TypeError("Invalid connection type")
                        pending.append(
                            row(
                                "connection_observations",
                                doc,
                                text(entry["id"]),
                                connection_id=entry["id"],
                                connection_name=estate_optional(entry["displayName"]),
                                gateway_id=estate_optional(entry.get("gatewayId")),
                                connectivity_type=text(entry["connectivityType"]),
                                privacy_level=estate_optional(
                                    entry.get("privacyLevel")
                                ),
                                connection_type=text(details["type"]),
                            )
                        )
                    status = "success" if manifest["complete"] else "partial"
                    issue = "caller_visible_only_endpoints_and_credentials_excluded"
                    unique = {}
                    for entry in pending:
                        key = entry["observation_id"]
                        if key in unique and unique[key] != entry:
                            raise ValueError("Conflicting connection identity")
                        unique[key] = entry
                else:
                    status, issue = "failed", "connection_resource_failed"
            except (ValueError, TypeError, KeyError, AttributeError):
                status, issue, pending = "invalid", "invalid_connection_metadata", []
        output["connection_observations"].extend(pending)
        output["estate_coverage"].append(
            row(
                "estate_coverage",
                run["doc"],
                "connections",
                analysis="connections",
                coverage_status=status,
                issue=issue,
                record_count=len(pending) if status in {"success", "partial"} else None,
            )
        )


def estate_inventory(documents, output):
    indexed = {doc["path"]: doc for doc in documents}
    for entry in output["inventory"]:
        source = entry["inventory_source"]
        if source not in {"legacy_items", "legacy_workspaces"}:
            continue
        doc = indexed[entry["source_path"]]
        if source == "legacy_items":
            output["catalog_items"].append(
                estate_row(
                    "catalog_items",
                    doc,
                    [entry["workspace_id"], entry["item_id"], entry["item_type"]],
                    workspace_id=entry["workspace_id"],
                    item_id=entry["item_id"],
                    item_type=entry["item_type"],
                    item_name=entry["item_name"],
                )
            )
        else:
            output["catalog_workspaces"].append(
                estate_row(
                    "catalog_workspaces",
                    doc,
                    entry["workspace_id"],
                    workspace_id=entry["workspace_id"],
                    workspace_name=entry["item_name"],
                )
            )
    for health in output["collection_health"]:
        if health["module"] not in {"items", "workspaces"}:
            continue
        doc = indexed.get(health["source_path"])
        if doc is None:
            continue
        output["estate_coverage"].append(
            estate_row(
                "estate_coverage",
                doc,
                "catalog_inventory",
                analysis="catalog",
                coverage_status=health["coverage_status"],
                source_date=estate_source_date(doc)
                if health["coverage_status"] != "invalid"
                else None,
                record_count=health["record_count"],
                issue="legacy_item_inventory_not_physical_schema",
            )
        )


def transform_estate(documents, runs, output):
    estate_connections(runs, output)
    estate_inventory(documents, output)
    for doc in documents:
        prefix = doc["path"].split("/")[0]
        domain = next(
            (
                name
                for name, sources in ESTATE_SOURCES.items()
                if prefix in sources and name != "connections"
            ),
            None,
        )
        if domain is None or (
            domain == "catalog" and not doc["path"].startswith("catalog/scans/")
        ):
            continue
        if domain == "gateways" and PurePosixPath(doc["path"]).name == "users.json":
            continue
        pending = {name: [] for name in ESTATE_DOMAINS[domain]}
        status, issue, count, source_date = (
            "legacy_unverified",
            "no_completion_manifest_or_exact_collection_time",
            None,
            None,
        )
        try:
            source_date = estate_source_date(doc)
            {
                "activity": estate_activity,
                "catalog": estate_catalog,
                "gateways": estate_gateways,
                "refresh": estate_refresh,
            }[domain](doc, pending)
            for entries in pending.values():
                unique = {}
                for entry in entries:
                    identity = entry["observation_id"]
                    if identity in unique and unique[identity] != entry:
                        raise ValueError("Conflicting raw identity")
                    unique[identity] = entry
            count = sum(len(entries) for entries in pending.values())
            if domain == "catalog":
                issue = "semantic_model_schema_only_physical_schemas_uncollected"
            for dataset, entries in pending.items():
                output[dataset].extend(entries)
        except (ValueError, TypeError, KeyError, AttributeError, IndexError):
            status, issue = "invalid", "invalid_or_unsupported_estate_artifact"
        output["estate_coverage"].append(
            estate_row(
                "estate_coverage",
                doc,
                domain,
                analysis=domain,
                coverage_status=status,
                source_date=source_date,
                record_count=count,
                issue=issue,
            )
        )


def preview_estate(spark_session, display_frame, tables, domain):
    from pyspark.sql import functions as F

    coverage = [
        entry for entry in tables["estate_coverage"] if entry["analysis"] == domain
    ]
    if not coverage:
        print(
            f"{domain}: no supported collected files. NOT COLLECTED, not successful zero."
        )
    else:
        display_frame(
            to_frame(spark_session, "estate_coverage", coverage)
            .groupBy("coverage_status", "issue")
            .count()
            .orderBy("coverage_status", "issue")
        )
    frames = {
        name: to_frame(spark_session, name, tables[name])
        for name in ESTATE_DOMAINS[domain]
    }
    print(
        "Charts use retained raw-file observations, not proof of current or complete estate coverage."
    )
    if domain == "activity":
        frame = frames["activity_events"]
        plot_counts(frame, "operation", "Retained activity observations by operation")
        display_frame(
            frame.groupBy("workload", "operation").agg(
                F.count("*").alias("observations"),
                F.countDistinct("event_id").alias("known_event_ids"),
                F.countDistinct("actor_key").alias("pseudonymous_actors"),
            )
        )
        display_frame(
            frame.groupBy(
                F.substring("event_time_utc", 1, 10).alias("event_date_utc")
            ).count()
        )
        print(
            "Actor keys are unsalted pseudonyms, not anonymous identities. UPNs, IPs and event payloads are excluded."
        )
    elif domain == "catalog":
        plot_counts(
            frames["catalog_items"], "item_type", "Scanner-returned item observations"
        )
        plot_counts(
            frames["catalog_models"],
            "schema_coverage",
            "Semantic-model schema availability",
        )
        plot_counts(frames["model_columns"], "data_type", "Returned model column types")
        display_frame(
            frames["catalog_models"]
            .select(
                "workspace_id",
                "item_id",
                "model_name",
                "schema_coverage",
                "table_count",
                "relationship_count",
                "source_date",
            )
            .limit(100)
        )
        display_frame(
            frames["model_columns"]
            .select(
                "workspace_id",
                "item_id",
                "table_name",
                "column_name",
                "data_type",
                "is_hidden",
            )
            .limit(100)
        )
        display_frame(
            frames["model_measures"]
            .select(
                "workspace_id",
                "item_id",
                "table_name",
                "measure_name",
                "expression_returned",
            )
            .limit(100)
        )
        display_frame(
            frames["model_relationships"]
            .select(
                "workspace_id",
                "item_id",
                "from_table",
                "from_column",
                "to_table",
                "to_column",
                "is_active",
            )
            .limit(100)
        )
        print(
            "Tables/columns/measures/relationships are available in managed model_* tables after publishing."
        )
        print(
            "Missing schema is not an empty model. Physical Lakehouse/Warehouse schemas and expression text are not exported."
        )
    elif domain == "connections":
        plot_counts(
            frames["connection_observations"],
            "connection_type",
            "Retained caller-visible connection types",
        )
        display_frame(
            frames["connection_observations"]
            .groupBy("connectivity_type", "privacy_level")
            .count()
        )
        display_frame(
            frames["connection_observations"]
            .filter(F.col("gateway_id").isNotNull())
            .groupBy("gateway_id")
            .agg(F.countDistinct("connection_id").alias("observed_connections"))
        )
        display_frame(
            frames["connection_observations"]
            .select(
                "connection_id",
                "connection_name",
                "connection_type",
                "connectivity_type",
                "gateway_id",
                "privacy_level",
                "observed_at_utc",
            )
            .limit(100)
        )
        print(
            "Caller-visible only. Failure/empty latest attempts do not prove older connections are current. Endpoints and credentials are excluded."
        )
    elif domain == "gateways":
        plot_counts(
            frames["gateway_datasources"],
            "datasource_type",
            "Retained datasource type observations",
        )
        plot_counts(
            frames["gateway_status"],
            "connectivity_observation",
            "Retained gateway connectivity observations",
        )
        display_frame(
            frames["gateway_status"]
            .groupBy("source_date", "status_code", "http_status")
            .count()
        )
        display_frame(
            frames["gateway_observations"]
            .select(
                "gateway_id",
                "gateway_name",
                "gateway_type",
                "source_date",
            )
            .limit(100)
        )
        display_frame(
            frames["gateway_datasources"]
            .select(
                "gateway_id",
                "datasource_id",
                "datasource_name",
                "datasource_type",
                "detail_returned",
                "source_date",
            )
            .limit(100)
        )
        paths = [entry["source_path"] for entry in coverage]
        for name in (
            "gateways.json",
            "datasources.json",
            "datasource_details.json",
            "status.json",
        ):
            if not any(path.endswith("/" + name) for path in paths):
                print(f"{name}: NOT COLLECTED; missing is not healthy/empty.")
        print(
            "Gateway API visibility is not tenant completeness or guaranteed VNet coverage. Status lacks exact observation time; users and connection details are excluded."
        )
    elif domain == "refresh":
        frame = frames["refresh_observations"]
        plot_counts(frame, "status", "Retained dataset refresh status observations")
        display_frame(
            frame.groupBy("status", "refresh_type").agg(
                F.count("*").alias("observations"),
                F.countDistinct("workspace_id", "item_id", "refresh_id").alias(
                    "known_refresh_keys"
                ),
                F.avg("duration_seconds").alias("mean_known_duration_seconds"),
            )
        )
        display_frame(
            frames["refreshable_observations"]
            .groupBy("schedule_enabled")
            .agg(
                F.count("*").alias("observations"),
                F.sum("refresh_failures").alias("reported_failure_observations"),
            )
        )
        display_frame(
            frame.filter(F.col("failure_code").isNotNull())
            .groupBy(
                "workspace_id",
                "item_id",
                "failure_code",
            )
            .count()
            .orderBy(F.desc("count"))
            .limit(100)
        )
        for source in ESTATE_SOURCES["refresh"]:
            if not any(entry["module"] == source for entry in coverage):
                print(f"{source}: NOT COLLECTED, not zero refreshes.")
        print(
            "Power BI semantic-model refresh only, not all Fabric jobs. Repeated history windows are retained observations; averages are not deduplicated run-level KPIs. Error messages are excluded."
        )


MODULE_DOMAINS = {
    "apps": ("apps",),
    "capacity": ("capacities", "capacity_metric_observations"),
    "directory": (
        "directory_users",
        "directory_groups",
        "group_memberships",
        "user_licenses",
        "subscribed_skus",
    ),
    "roles": ("workspace_role_scopes", "workspace_role_assignments"),
    "settings": (
        "tenant_settings",
        "delegated_setting_scopes",
        "delegated_setting_overrides",
        "tenant_setting_groups",
    ),
    "domains": ("domains", "domain_workspaces"),
    "git": ("workspace_git_connections",),
    "tags": ("tags",),
    "sharing": ("organization_shares", "publish_to_web_shares"),
    "deployments": (
        "deployment_pipelines",
        "deployment_operation_resources",
        "deployment_operations",
    ),
}
MODULE_SOURCES = {
    "apps": ("catalog",),
    "capacity": ("capacity", "capacity_metrics"),
    "directory": ("graph",),
    "roles": ("roles",),
    "settings": ("tenant", "delegatedsettings"),
    "domains": ("domains",),
    "git": ("gitconnections",),
    "tags": ("tags",),
    "sharing": ("widelysharedlinks",),
    "deployments": ("fabricdeploymentpipelineoperations",),
}
MODULE_LIMITS = {
    "apps": "App inventory only; app audiences/access and complete paging are not established.",
    "capacity": "Metrics app extraction is opt-in. Values/units are unchanged; naive timepoints have unknown UTC. Snapshots are not additive KPIs.",
    "directory": "Groups/members are optional and capped; membership is transitive USERS only. IDs are not anonymous; names, UPNs and mail are excluded.",
    "roles": "Observed workspace roles, not item grants or effective access. Omitted users are unknown, not zero assignments.",
    "settings": "Observed settings/group rules, not computed effective policy or precedence. Missing override collections stay unknown.",
    "domains": "The collector omits domains with no workspaces. Derived domain inventory is therefore incomplete.",
    "git": "Observed repository/branch metadata, not source code or sync health proof. Owners/directories are pseudonyms; URLs/credentials are excluded.",
    "tags": "Tag definitions only; item/tag assignments are not collected.",
    "sharing": "Observed organization/public exposures, not complete access or verified live links. Sharer identities and link URLs are excluded.",
    "deployments": "Caller-visible pipelines; operation history needs pipeline Admin and exposes up to 20 recent deployments. Polling gaps cannot be recovered.",
}
MODULE_ARTIFACTS = {
    "apps": ("apps.json",),
    "directory": (
        "users.json",
        "groups.json",
        "groupmembers.json",
        "userlicenses.json",
        "subscribedskus.json",
    ),
    "roles": (),
    "capacity": (),
    "deployments": (),
    "settings": (
        "tenant-settings.json",
        "capacities.json",
        "domains.json",
        "workspaces.json",
    ),
    "domains": ("domains.json",),
    "git": ("gitconnections.json",),
    "tags": ("tags.json",),
    "sharing": (
        "widelysharedlinks_organization.json",
        "widelysharedlinks_publishedtoweb.json",
    ),
}
METRIC_NAMES = {
    "CU limit",
    "SKU CU by timepoint %",
    "Background billable CU %",
    "Interactive billable CU %",
    "SKU CU by timepoint",
    "Background non billable CU %",
    "Interactive non billable CU %",
    "Carry over add %",
    "Carry over burndown %",
    "Cumulative carry over %",
    "Expected burndown in minutes",
}


def module_source_date(doc):
    parts = PurePosixPath(doc["path"]).parts
    offset = 2 if parts[0] in {"catalog", "delegatedsettings"} else 1
    if parts[0] == "catalog" and parts[1] != "snapshots":
        raise ValueError("Not an app snapshot")
    if parts[0] == "delegatedsettings" and parts[1] not in {
        "capacities",
        "domains",
        "workspaces",
    }:
        raise ValueError("Unknown delegated scope")
    expected = 6 if parts[0] == "fabricdeploymentpipelineoperations" else offset + 4
    if len(parts) != expected:
        raise ValueError("Unexpected module artifact path")
    date = "/".join(parts[offset : offset + 3])
    if not re.fullmatch(r"\d{4}/\d{2}/\d{2}", date):
        raise ValueError("Unexpected module date partition")
    return datetime.strptime(date, "%Y/%m/%d").strftime("%Y-%m-%d")


def module_row(dataset, doc, key, **values):
    return estate_row(dataset, doc, key, source_date=module_source_date(doc), **values)


def module_object(value):
    if value is None:
        return {}
    if not isinstance(value, dict):
        raise TypeError("Invalid metadata object")
    return value


def module_strings(value):
    if value is None:
        return None
    if not isinstance(value, list) or any(not isinstance(item, str) for item in value):
        raise TypeError("Invalid string collection")
    return canonical(value)


def module_key(value):
    if value is None:
        return None
    value = text(value)
    try:
        return str(UUID(value))
    except ValueError:
        return "sha256:" + sha256(value.casefold().encode())


def module_time(value):
    if value is None:
        return None
    parsed = datetime.fromisoformat(text(value).replace("Z", "+00:00"))
    return utc(value) if parsed.tzinfo is not None else None


def module_duration(start, end):
    if start is None or end is None:
        return None
    delta = (
        datetime.fromisoformat(end.replace("Z", "+00:00"))
        - datetime.fromisoformat(start.replace("Z", "+00:00"))
    ).total_seconds()
    return delta if delta >= 0 else None


def module_apps(doc, pending):
    if PurePosixPath(doc["path"]).name != "apps.json":
        raise ValueError("Unexpected app file")
    for item in estate_records(doc):
        identity = text(item["id"])
        pending["apps"].append(
            module_row(
                "apps",
                doc,
                identity,
                app_id=identity,
                app_name=estate_optional(item.get("name")),
                workspace_id=estate_optional(item.get("workspaceId")),
                last_update_utc=module_time(item.get("lastUpdate")),
            )
        )


def module_capacity(doc, pending):
    if doc["path"].split("/")[0] == "capacity":
        if not re.fullmatch(
            r"\d{8}_\d+\.capacity\.json", PurePosixPath(doc["path"]).name
        ):
            raise ValueError("Unexpected capacity file")
        for item in estate_records(doc):
            identity = text(item["id"])
            pending["capacities"].append(
                module_row(
                    "capacities",
                    doc,
                    identity,
                    capacity_id=identity,
                    capacity_name=estate_optional(item.get("displayName")),
                    sku=estate_optional(item.get("sku")),
                    state=estate_optional(item.get("state")),
                    region=estate_optional(item.get("region")),
                )
            )
        return
    value = doc["value"]
    if not isinstance(value, dict) or doc["issue"]:
        raise TypeError("Invalid metrics envelope")
    capacity = text(value["capacityId"])
    UUID(capacity)
    group = text(value["group"])
    date = datetime.strptime(text(value["date"]), "%Y-%m-%d").strftime("%Y-%m-%d")
    filename = f"{module_source_date(doc).replace('-', '')}_{capacity}_{group}_{date.replace('-', '')}.capacity_metrics.json"
    if (
        group not in {"utilization", "carryover"}
        or PurePosixPath(doc["path"]).name != filename
    ):
        raise ValueError("Metrics identity does not match artifact")
    for item in records(value["rows"]):
        timepoint = item.get("Timepoints[Timepoint]", item.get("[Timepoint]"))
        raw_time = text(timepoint)
        time_utc = module_time(raw_time)
        metrics = [name for name in sorted(METRIC_NAMES) if "[" + name + "]" in item]
        if not metrics:
            raise ValueError("No supported returned metrics")
        for name in metrics:
            number = item["[" + name + "]"]
            if number is not None and (
                type(number) not in (int, float) or not math.isfinite(number)
            ):
                raise ValueError("Invalid metric value")
            pending["capacity_metric_observations"].append(
                module_row(
                    "capacity_metric_observations",
                    doc,
                    [raw_time, name],
                    capacity_id=capacity,
                    sku=estate_optional(value.get("sku")),
                    metric_group=group,
                    metric_date=date,
                    timepoint_raw=raw_time,
                    timepoint_utc=time_utc,
                    metric_name=name,
                    metric_value=float(number) if number is not None else None,
                )
            )


def module_directory(doc, pending):
    name = PurePosixPath(doc["path"]).name
    if name not in MODULE_ARTIFACTS["directory"]:
        raise ValueError("Unexpected directory file")
    for item in estate_records(doc):
        if name == "users.json":
            identity = module_key(text(item["id"]))
            licenses = estate_optional_records(item.get("assignedLicenses"))
            pending["directory_users"].append(
                module_row(
                    "directory_users",
                    doc,
                    identity,
                    principal_id=identity,
                    account_enabled=estate_optional(item.get("accountEnabled"), bool),
                    user_type=estate_optional(item.get("userType")),
                    assigned_license_count=len(licenses)
                    if licenses is not None
                    else None,
                )
            )
        elif name == "groups.json":
            identity = module_key(text(item["id"]))
            pending["directory_groups"].append(
                module_row(
                    "directory_groups",
                    doc,
                    identity,
                    group_id=identity,
                    security_enabled=estate_optional(item.get("securityEnabled"), bool),
                    mail_enabled=estate_optional(item.get("mailEnabled"), bool),
                    group_types_json=module_strings(item.get("groupTypes")),
                    is_audience_group=estate_optional(
                        item.get("isPowerBIAudienceGroup"), bool
                    ),
                )
            )
        elif name == "groupmembers.json":
            group, principal = (
                module_key(text(item["groupId"])),
                module_key(text(item["memberId"])),
            )
            pending["group_memberships"].append(
                module_row(
                    "group_memberships",
                    doc,
                    [group, principal],
                    group_id=group,
                    principal_id=principal,
                    membership_kind="transitive_user",
                    is_audience_group=estate_optional(
                        item.get("isPowerBIAudienceGroup"), bool
                    ),
                )
            )
        elif name == "userlicenses.json":
            principal, sku = module_key(text(item["userId"])), text(item["skuId"])
            pending["user_licenses"].append(
                module_row(
                    "user_licenses",
                    doc,
                    [principal, sku],
                    principal_id=principal,
                    sku_id=sku,
                    disabled_plans_json=module_strings(item.get("disabledPlans")),
                )
            )
        else:
            sku = text(item["skuId"])
            units = module_object(item.get("prepaidUnits"))
            pending["subscribed_skus"].append(
                module_row(
                    "subscribed_skus",
                    doc,
                    sku,
                    sku_id=sku,
                    sku_part_number=estate_optional(item.get("skuPartNumber")),
                    capability_status=estate_optional(item.get("capabilityStatus")),
                    consumed_units=estate_count(item.get("consumedUnits")),
                    enabled_units=estate_count(units.get("enabled")),
                    suspended_units=estate_count(units.get("suspended")),
                    warning_units=estate_count(units.get("warning")),
                )
            )


def module_roles(doc, pending):
    if not re.fullmatch(r"\d{8}_\d+\.roles\.json", PurePosixPath(doc["path"]).name):
        raise ValueError("Unexpected roles file")
    for workspace in estate_records(doc):
        identity = text(workspace["id"])
        users = estate_optional_records(workspace.get("users"))
        pending["workspace_role_scopes"].append(
            module_row(
                "workspace_role_scopes",
                doc,
                identity,
                workspace_id=identity,
                roles_returned=users is not None,
                assignment_count=len(users) if users is not None else None,
            )
        )
        for index, user in enumerate(users or []):
            principal = module_key(user.get("graphId") or user.get("identifier"))
            role = estate_optional(user.get("groupUserAccessRight"))
            pending["workspace_role_assignments"].append(
                module_row(
                    "workspace_role_assignments",
                    doc,
                    [identity, principal, role, index],
                    workspace_id=identity,
                    principal_id=principal,
                    principal_type=estate_optional(user.get("principalType")),
                    workspace_role=role,
                )
            )


def module_settings(doc, pending):
    tenant = doc["path"].split("/")[0] == "tenant"
    scope = "tenant" if tenant else doc["path"].split("/")[1]
    if tenant:
        if PurePosixPath(doc["path"]).name != "tenant-settings.json" or doc["issue"]:
            raise ValueError("Unexpected tenant settings file")
        root = module_object(doc["value"])
        settings = records(root["tenantSettings"])
        parents = [(None, settings)]
    else:
        if PurePosixPath(doc["path"]).name != scope + ".json":
            raise ValueError("Unexpected delegated settings file")
        parents = []
        for entity in estate_records(doc):
            identity = text(entity["id"])
            if (
                "tenantSettings" in entity
                and "overrides" in entity
                and entity["tenantSettings"] != entity["overrides"]
            ):
                raise ValueError("Conflicting override collections")
            settings = estate_optional_records(
                entity.get("tenantSettings", entity.get("overrides"))
            )
            parents.append((identity, settings))
            pending["delegated_setting_scopes"].append(
                module_row(
                    "delegated_setting_scopes",
                    doc,
                    identity,
                    scope_type=scope,
                    entity_id=identity,
                    settings_returned=settings is not None,
                    setting_count=len(settings) if settings is not None else None,
                )
            )
    for identity, settings in parents:
        for setting in settings or []:
            name = text(setting["settingName"])
            counts = {}
            for field, rule in (
                ("enabledSecurityGroups", "enabled"),
                ("excludedSecurityGroups", "excluded"),
            ):
                groups = setting.get(field)
                if groups is not None and not isinstance(groups, list):
                    raise TypeError("Invalid setting group collection")
                counts[rule] = len(groups) if groups is not None else None
                for group in groups or []:
                    group_id = (
                        group
                        if isinstance(group, str)
                        else module_object(group).get("graphId")
                        or module_object(group).get("id")
                    )
                    group_key = module_key(text(group_id))
                    pending["tenant_setting_groups"].append(
                        module_row(
                            "tenant_setting_groups",
                            doc,
                            [scope, identity, name, rule, group_key],
                            scope_type=scope,
                            entity_id=identity,
                            setting_name=name,
                            group_id=group_key,
                            group_rule=rule,
                        )
                    )
            values = {
                "setting_name": name,
                "title": estate_optional(setting.get("title")),
                "setting_group": estate_optional(setting.get("tenantSettingGroup")),
                "enabled": estate_optional(setting.get("enabled"), bool),
                "enabled_group_count": counts["enabled"],
                "excluded_group_count": counts["excluded"],
            }
            if tenant:
                pending["tenant_settings"].append(
                    module_row(
                        "tenant_settings",
                        doc,
                        name,
                        **values,
                        can_specify_groups=estate_optional(
                            setting.get("canSpecifySecurityGroups"), bool
                        ),
                    )
                )
            else:
                pending["delegated_setting_overrides"].append(
                    module_row(
                        "delegated_setting_overrides",
                        doc,
                        [identity, name],
                        **values,
                        scope_type=scope,
                        entity_id=identity,
                        delegated_from=estate_optional(setting.get("delegatedFrom")),
                    )
                )


def module_domains(doc, pending):
    if PurePosixPath(doc["path"]).name != "domains.json":
        raise ValueError("Unexpected domains file")
    for item in estate_records(doc):
        domain, workspace = text(item["id"]), text(item["workspaceId"])
        pending["domains"].append(
            module_row(
                "domains",
                doc,
                domain,
                domain_id=domain,
                domain_name=estate_optional(item.get("displayName")),
                parent_domain_id=estate_optional(item.get("parentDomainId")),
            )
        )
        pending["domain_workspaces"].append(
            module_row(
                "domain_workspaces",
                doc,
                [domain, workspace],
                domain_id=domain,
                workspace_id=workspace,
                workspace_name=estate_optional(item.get("workspace")),
            )
        )


def module_git(doc, pending):
    if PurePosixPath(doc["path"]).name != "gitconnections.json":
        raise ValueError("Unexpected Git file")
    for item in estate_records(doc):
        identity = text(item["workspaceId"])
        provider, sync = (
            module_object(item.get("gitProviderDetails")),
            module_object(item.get("gitSyncDetails")),
        )
        owner, directory = (
            estate_optional(provider.get("ownerName")),
            estate_optional(provider.get("directoryName")),
        )
        pending["workspace_git_connections"].append(
            module_row(
                "workspace_git_connections",
                doc,
                identity,
                workspace_id=identity,
                provider_type=estate_optional(provider.get("gitProviderType")),
                owner_key=sha256(owner.encode()) if owner is not None else None,
                organization_name=estate_optional(provider.get("organizationName")),
                project_name=estate_optional(provider.get("projectName")),
                repository_name=estate_optional(provider.get("repositoryName")),
                branch_name=estate_optional(provider.get("branchName")),
                directory_key=sha256(directory.encode())
                if directory is not None
                else None,
                connection_state=estate_optional(item.get("gitConnectionState")),
                last_sync_utc=module_time(sync.get("lastSyncTime")),
            )
        )


def module_tags(doc, pending):
    if PurePosixPath(doc["path"]).name != "tags.json":
        raise ValueError("Unexpected tags file")
    for item in estate_records(doc):
        identity, scope = text(item["id"]), module_object(item.get("scope"))
        pending["tags"].append(
            module_row(
                "tags",
                doc,
                identity,
                tag_id=identity,
                tag_name=estate_optional(item.get("displayName")),
                scope_type=estate_optional(scope.get("type")),
                domain_id=estate_optional(scope.get("domainId")),
            )
        )


def module_sharing(doc, pending):
    name = PurePosixPath(doc["path"]).name
    if name not in MODULE_ARTIFACTS["sharing"]:
        raise ValueError("Unexpected sharing file")
    dataset = (
        "publish_to_web_shares" if "publishedtoweb" in name else "organization_shares"
    )
    for item in estate_records(doc):
        identity = text(item["artifactId"])
        access, share = (
            estate_optional(item.get("accessRight")),
            estate_optional(item.get("shareType")),
        )
        pending[dataset].append(
            module_row(
                dataset,
                doc,
                [identity, access, share],
                artifact_id=identity,
                artifact_type=estate_optional(item.get("artifactType")),
                access_right=access,
                share_type=share,
            )
        )


def module_append(output, pending):
    prepared = {}
    for dataset, entries in pending.items():
        unique = {}
        for entry in entries:
            identity = entry["observation_id"]
            if identity in unique and unique[identity] != entry:
                raise ValueError("Conflicting normalized identity")
            unique[identity] = entry
        prepared[dataset] = list(unique.values())
    for dataset, entries in prepared.items():
        output[dataset].extend(entries)
    return sum(len(entries) for entries in prepared.values())


def module_deployments(runs, output):
    for run in runs.values():
        if run["module"] != "fabricdeploymentpipelineoperations":
            continue
        status, issue, count = "invalid", "invalid_deployment_run", None
        manifest = run["manifest"]
        pending = {name: [] for name in MODULE_DOMAINS["deployments"]}
        if run["valid"] and not manifest["finished"]:
            status, issue = "unfinished", "unfinished_deployment_run"
        elif run["valid"]:
            try:
                inventory = [
                    doc
                    for doc in run["resources"]
                    if PurePosixPath(doc["path"]).name == "pipelines.json"
                ]
                if len(inventory) != 1:
                    raise ValueError("Missing pipeline inventory")
                inventory = inventory[0]
                resource = inventory["value"]
                if (
                    resource["requestPath"] != "deploymentPipelines"
                    or resource["parent"]
                ):
                    raise ValueError("Wrong pipeline inventory endpoint")
                pipelines = {}
                if resource["complete"]:
                    for item in records(resource["records"]):
                        identity = str(UUID(text(item["id"])))
                        if identity in pipelines:
                            raise ValueError("Duplicate pipeline")
                        pipelines[identity] = item
                        pending["deployment_pipelines"].append(
                            row(
                                "deployment_pipelines",
                                inventory,
                                identity,
                                pipeline_id=identity,
                                pipeline_name=text(item["displayName"]),
                                source_date=module_source_date(inventory),
                            )
                        )
                expected = {
                    "pipelines.json",
                    *(identity + ".json" for identity in pipelines),
                }
                if {
                    PurePosixPath(doc["path"]).name for doc in run["resources"]
                } != expected:
                    raise ValueError("Missing or unexpected pipeline resource")
                for doc in run["resources"]:
                    if doc is inventory:
                        continue
                    value = doc["value"]
                    identity = PurePosixPath(doc["path"]).stem
                    if (
                        value["parent"] != {"deploymentPipeline": pipelines[identity]}
                        or value["requestPath"]
                        != f"deploymentPipelines/{identity}/operations"
                    ):
                        raise ValueError("Operation parent or endpoint mismatch")
                    values = records(value["records"]) if value["complete"] else []
                    pending["deployment_operation_resources"].append(
                        row(
                            "deployment_operation_resources",
                            doc,
                            identity,
                            pipeline_id=identity,
                            source_date=module_source_date(doc),
                            coverage_status="success"
                            if value["complete"]
                            else "failed",
                            resource_outcome="success"
                            if value["complete"]
                            else "failed",
                            record_count=len(values) if value["complete"] else None,
                        )
                    )
                    for item in values:
                        operation = text(item["id"])
                        start, end = (
                            module_time(item.get("executionStartTime")),
                            module_time(item.get("executionEndTime")),
                        )
                        pending["deployment_operations"].append(
                            row(
                                "deployment_operations",
                                doc,
                                operation,
                                pipeline_id=identity,
                                operation_id=operation,
                                operation_type=text(item["type"]),
                                status=text(item["status"]),
                                last_updated_utc=module_time(
                                    item.get("lastUpdatedTime")
                                ),
                                execution_start_utc=start,
                                execution_end_utc=end,
                                duration_seconds=module_duration(start, end),
                                source_stage_id=estate_optional(
                                    item.get("sourceStageId")
                                ),
                                target_stage_id=estate_optional(
                                    item.get("targetStageId")
                                ),
                                source_date=module_source_date(doc),
                            )
                        )
                count = module_append(output, pending)
                status = "success" if manifest["complete"] else "partial"
                issue = MODULE_LIMITS["deployments"]
                if not resource["complete"]:
                    count, status = None, "failed"
            except (ValueError, KeyError, TypeError, AttributeError, IndexError):
                status, issue, count = (
                    "invalid",
                    "invalid_deployment_metadata_or_scope",
                    None,
                )
        output["estate_coverage"].append(
            row(
                "estate_coverage",
                run["doc"],
                "deployments",
                analysis="deployments",
                coverage_status=status,
                record_count=count,
                issue=issue,
            )
        )


def transform_modules(documents, runs, output):
    module_deployments(runs, output)
    parsers = {
        "apps": module_apps,
        "capacity": module_capacity,
        "directory": module_directory,
        "roles": module_roles,
        "settings": module_settings,
        "domains": module_domains,
        "git": module_git,
        "tags": module_tags,
        "sharing": module_sharing,
    }
    for doc in documents:
        prefix = doc["path"].split("/")[0]
        domain = next(
            (
                name
                for name, sources in MODULE_SOURCES.items()
                if prefix in sources and name != "deployments"
            ),
            None,
        )
        if domain is None or (
            domain == "apps" and not doc["path"].startswith("catalog/snapshots/")
        ):
            continue
        status, count, date, issue = (
            "legacy_unverified",
            None,
            None,
            MODULE_LIMITS[domain],
        )
        pending = {name: [] for name in MODULE_DOMAINS[domain]}
        try:
            date = module_source_date(doc)
            parsers[domain](doc, pending)
            count = module_append(output, pending)
            if isinstance(doc["value"], dict) and doc["value"].get("continuationToken"):
                issue = "source_continuation_present_completeness_unverified"
        except (
            ValueError,
            KeyError,
            TypeError,
            AttributeError,
            IndexError,
            OverflowError,
        ):
            status, issue = "invalid", "invalid_or_unavailable_module_metadata"
        output["estate_coverage"].append(
            estate_row(
                "estate_coverage",
                doc,
                domain,
                analysis=domain,
                source_date=date,
                coverage_status=status,
                record_count=count,
                issue=issue,
            )
        )


def module_capacity_chart(frame):
    import matplotlib.pyplot as plt
    from pyspark.sql import functions as F

    chosen = (
        frame.filter(F.col("metric_value").isNotNull())
        .orderBy(
            F.col("source_date").desc(),
            F.col("metric_date").desc(),
            "capacity_id",
            "metric_group",
            "metric_name",
        )
        .limit(1)
        .collect()
    )
    if not chosen:
        print("No numeric capacity metrics collected; extraction is opt-in.")
        return
    selected = chosen[0]
    fields = (
        "source_date",
        "metric_date",
        "capacity_id",
        "metric_group",
        "metric_name",
    )
    for key in fields:
        frame = frame.filter(F.col(key) == selected[key])
    total = frame.count()
    points = (
        frame.select("timepoint_raw", "metric_value")
        .orderBy("timepoint_raw")
        .limit(200)
        .toPandas()
    )
    axis = points.plot(
        x="timepoint_raw", y="metric_value", legend=False, title=selected["metric_name"]
    )
    axis.set_xlabel("Provider timepoint (timezone may be unknown)")
    axis.set_ylabel("Unscaled provider value")
    print(
        {key: selected[key] for key in fields},
        f"showing {len(points)}/{total} observations",
    )
    plt.tight_layout()
    plt.show()


def preview_modules(spark_session, display_frame, tables, domain):
    from pyspark.sql import functions as F

    coverage = [
        entry for entry in tables["estate_coverage"] if entry["analysis"] == domain
    ]
    print(MODULE_LIMITS[domain])
    print(
        "Retained raw-file observations, not current tenant-wide or effective-state claims."
    )
    if not coverage:
        print("NOT COLLECTED: no supported source files, not successful zero.")
    else:
        display_frame(
            to_frame(spark_session, "estate_coverage", coverage)
            .groupBy("coverage_status", "issue")
            .count()
        )
    for name in MODULE_ARTIFACTS[domain]:
        if not any(entry["source_path"].endswith("/" + name) for entry in coverage):
            print(
                f"{name}: NOT COLLECTED; optional, failed or unavailable is not empty success."
            )
    frames = {
        name: to_frame(spark_session, name, tables[name])
        for name in MODULE_DOMAINS[domain]
    }
    categories = {
        "apps": ("apps", "workspace_id"),
        "capacity": ("capacities", "sku"),
        "directory": ("directory_users", "user_type"),
        "roles": ("workspace_role_assignments", "workspace_role"),
        "settings": ("tenant_settings", "enabled"),
        "domains": ("domain_workspaces", "domain_id"),
        "git": ("workspace_git_connections", "connection_state"),
        "tags": ("tags", "scope_type"),
        "sharing": ("organization_shares", "artifact_type"),
        "deployments": ("deployment_operations", "status"),
    }
    dataset, category = categories[domain]
    plotted = frames[dataset]
    if domain == "deployments":
        plotted = latest_frame(plotted, ["pipeline_id", "operation_id"])
    plot_counts(plotted, category, domain + ": retained observations")
    for name, frame in frames.items():
        print(
            name + ": at most 100 metadata rows; history is not a current-state union"
        )
        display_frame(
            frame.orderBy(
                F.col("observed_at_utc").desc_nulls_last(), F.col("source_path").desc()
            ).limit(100)
        )
    if domain == "capacity":
        metrics = frames["capacity_metric_observations"]
        display_frame(
            metrics.groupBy("capacity_id", "metric_group", "metric_name").agg(
                F.count("*").alias("observations"),
                F.count("metric_value").alias("known_values"),
                F.avg("metric_value").alias("mean_provider_value"),
            )
        )
        module_capacity_chart(metrics)
    if domain == "sharing":
        plot_counts(
            frames["publish_to_web_shares"],
            "artifact_type",
            "Retained publish-to-web observations",
        )
```
